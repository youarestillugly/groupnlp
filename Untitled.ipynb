## Configuration A: Standalone LLM Evaluation

### Objective

This configuration evaluates the Qwen2.5-0.5B-Instruct language model without using an external document retrieval system.

The model receives each evaluation question directly and generates an answer using its internal knowledge.

### Purpose

This experiment establishes the baseline performance of the language model. Its results will later be compared with BM25, Dense Retrieval, and Hybrid Retrieval configurations.

### Evaluation procedure

1. Load the existing Qwen2.5-0.5B-Instruct GGUF model.
2. Use the same model configuration as the original research assistant.
3. Submit all 80 evaluation questions directly to the model.
4. Record generated answers and response times.
5. Save the results for subsequent evaluation.

### 2.1 Checking Model Requirements

Before loading the language model, this section verifies that the llama-cpp-python library is installed and that the previously used GGUF model file is accessible.

This prevents unnecessary downloads and helps maintain consistency with the original research assistant.

```python
# Import required libraries
from pathlib import Path
import importlib.util

# Define the existing model path
MODEL_PATH = Path(
    "/Users/macbookair/DAT401_Research_Assistant/models/"
    "qwen2.5-0.5b-instruct-q4_k_m.gguf"
)

# Check whether llama-cpp-python is installed
llama_installed = (
    importlib.util.find_spec("llama_cpp") is not None
)

print("LLAMA.CPP INSTALLATION CHECK")
print("-" * 40)
print("llama-cpp-python installed:", llama_installed)

# Check whether the model file exists
print("\nMODEL FILE CHECK")
print("-" * 40)
print("Model path:", MODEL_PATH)
print("Model file exists:", MODEL_PATH.exists())

# Display model file size if available
if MODEL_PATH.exists():
    size_mb = MODEL_PATH.stat().st_size / (1024 ** 2)
    print(f"Model file size: {size_mb:.2f} MB")
else:
    print("Model file was not found.")
```

### 2.2 Loading the Standalone Language Model

This section loads the Qwen2.5-0.5B-Instruct model using the llama-cpp-python library.

The model is loaded from the existing GGUF file used in the original Research Assistant.

The same model configuration is maintained to ensure consistency between the original system and the evaluation experiments.

The loaded model will serve as the standalone baseline (Configuration A), without using document retrieval.

```python
# Import the Llama class from llama-cpp-python
from llama_cpp import Llama

# Load the existing Qwen2.5-0.5B-Instruct GGUF model
llm = Llama(
    model_path=str(MODEL_PATH),
    n_ctx=2048,
    n_batch=128,
    n_threads=4,
    verbose=False
)

# Confirm that the model has loaded successfully
print("Standalone LLM loaded successfully!")
print("Model:", MODEL_PATH.name)
print("Runtime:", type(llm).__name__)
print("Context size:", llm.n_ctx())
print("Batch size:", llm.n_batch)
print("CPU threads:", llm.n_threads)
```

```python
# Test the standalone model with a simple question

test_prompt = """<|im_start|>system
You are a research assistant. Answer the question clearly and concisely.
<|im_end|>
<|im_start|>user
What is a brain tumour?
<|im_end|>
<|im_start|>assistant
"""

# Generate a response
test_response = llm(
    test_prompt,
    max_tokens=150,
    temperature=0.1,
    stop=["<|im_end|>", "<|endoftext|>"]
)

# Extract the generated answer
test_answer = test_response["choices"][0]["text"].strip()

print("TEST QUESTION:")
print("What is a brain tumour?")

print("\nGENERATED ANSWER:")
print(test_answer)
```

### 2.3 Preparation of Evaluation Questions

This section prepares the verified evaluation dataset for the standalone LLM experiment.

The 80 questions are loaded from the saved evaluation dataset and organised for systematic evaluation.

Each question will be submitted directly to Qwen2.5-0.5B-Instruct without document retrieval.

The original question IDs, categories, expected answers, and answerability labels will be preserved to support consistent evaluation and comparison across all four system configurations.

```python
#Prepare Standalone Evalutation Dataset
# Import pandas for dataset handling
import pandas as pd

# Load the verified evaluation dataset
evaluation_df = pd.read_csv(
    "evaluation/evaluation_dataset_80_questions.csv"
)

# Display the dataset information
print("STANDALONE EVALUATION DATASET")
print("-" * 50)

print("Total questions:", len(evaluation_df))

print("\nQuestions by category:")
print(evaluation_df["category"].value_counts())

# Check the first five questions
print("\nFirst five evaluation questions:")
display(
    evaluation_df[
        ["question_id", "category", "question", "answerable"]
    ].head()
)

# Create a separate copy for Configuration A
config_a_df = evaluation_df.copy()

print("\nConfiguration A dataset prepared successfully!")
print("Total questions ready:", len(config_a_df))
```

### 2.4 Standalone LLM Answer Generation

In this experiment, the 80 evaluation questions are submitted directly to Qwen2.5-0.5B-Instruct without external document retrieval.

The model generates answers using its internal knowledge.

For each question, the generated answer and response time are recorded alongside the question ID, category, gold answer, and answerability label.

The generated results will be used to evaluate the standalone model and compare its performance with the retrieval-augmented configurations.

```python
# Import required libraries
import time
import pandas as pd

# Select the first three questions for testing
test_df = config_a_df.head(3).copy()

# Create a list to store generated results
test_results = []

# Generate answers for the first three questions
for _, row in test_df.iterrows():

    # Extract the evaluation question
    question = row["question"]

    # Prepare the Qwen instruction prompt
    prompt = f"""<|im_start|>system
You are a research assistant specialising in brain tumour research.
Answer the question clearly and accurately using your internal knowledge.
If you do not know the answer, say that you do not know.
Do not invent facts or citations.
<|im_end|>
<|im_start|>user
{question}
<|im_end|>
<|im_start|>assistant
"""

    # Record the generation start time
    start_time = time.time()

    # Generate an answer without document retrieval
    response = llm(
        prompt,
        max_tokens=250,
        temperature=0.1,
        stop=["<|im_end|>", "<|endoftext|>"]
    )

    # Calculate response time
    response_time = time.time() - start_time

    # Extract the generated answer
    generated_answer = response["choices"][0]["text"].strip()

    # Store the question and generation results
    test_results.append({
        "question_id": row["question_id"],
        "category": row["category"],
        "question": question,
        "gold_answer": row["gold_answer"],
        "answerable": row["answerable"],
        "supporting_sources": row["supporting_sources"],
        "generated_answer": generated_answer,
        "response_time": round(response_time, 3)
    })

    # Display the generated result
    print(f"\nQuestion ID: {row['question_id']}")
    print("Generated Answer:", generated_answer)
    print(f"Response Time: {response_time:.3f} seconds")

# Convert the generated results into a DataFrame
config_a_test_results = pd.DataFrame(test_results)

print("\nThree-question generation test completed successfully.")
print("Total generated answers:", len(config_a_test_results))
```

### 2.4.1 Adjustment of Answer Generation Length

The initial generation test showed that some answers were truncated because of the token limit.

To reduce premature answer termination, the maximum generation length is increased from 250 to 400 tokens.

The same three questions are tested again before proceeding with the complete evaluation dataset.

```python
# Repeat the generation test with a higher token limit

revised_test_results = []

for _, row in test_df.iterrows():

    # Extract the question
    question = row["question"]

    # Prepare the prompt
    prompt = f"""<|im_start|>system
You are a research assistant specialising in brain tumour research.
Answer the question clearly and accurately using your internal knowledge.
If you do not know the answer, say that you do not know.
Do not invent facts or citations.
<|im_end|>
<|im_start|>user
{question}
<|im_end|>
<|im_start|>assistant
"""

    # Start measuring response time
    start_time = time.time()

    # Generate the answer using a 400-token limit
    response = llm(
        prompt,
        max_tokens=400,
        temperature=0.1,
        stop=["<|im_end|>", "<|endoftext|>"]
    )

    # Calculate response time
    response_time = time.time() - start_time

    # Extract the generated answer
    generated_answer = response["choices"][0]["text"].strip()

    # Store the result
    revised_test_results.append({
        "question_id": row["question_id"],
        "generated_answer": generated_answer,
        "response_time": round(response_time, 3),
        "finish_reason": response["choices"][0].get("finish_reason")
    })

    print(f"\nQuestion ID: {row['question_id']}")
    print("Generated Answer:", generated_answer)
    print("Finish reason:", response["choices"][0].get("finish_reason"))
    print(f"Response time: {response_time:.3f} seconds")

# Create a DataFrame of revised results
config_a_revised_test = pd.DataFrame(revised_test_results)

print("\nRevised generation test completed.")
```

### 2.5 Full Standalone LLM Evaluation

After confirming that the generation procedure works correctly, the standalone language model is evaluated using all 80 verified questions.

Each question is submitted directly to Qwen2.5-0.5B-Instruct without document retrieval.

The experiment records the generated answer, response time, question category, gold answer, answerability label, and supporting sources.

The results are saved in CSV format for subsequent evaluation and comparison with the other three system configurations.

The generation settings remain consistent with the preliminary test:
- Maximum generation tokens: 400
- Temperature: 0.1
- Document retrieval: Disabled

```python
# Import required libraries
import time
import pandas as pd
from pathlib import Path

# Create the evaluation output directory
EVALUATION_DIR = Path("evaluation")
EVALUATION_DIR.mkdir(exist_ok=True)

# Define the output file
CONFIG_A_PATH = (
    EVALUATION_DIR / "configuration_A_standalone_results.csv"
)

# Load previously completed results if the experiment was interrupted
if CONFIG_A_PATH.exists():
    config_a_results = pd.read_csv(CONFIG_A_PATH)
    completed_ids = set(config_a_results["question_id"].astype(str))
    print("Existing results found:", len(completed_ids))
else:
    config_a_results = pd.DataFrame()
    completed_ids = set()

# Count questions remaining
remaining_questions = config_a_df[
    ~config_a_df["question_id"].astype(str).isin(completed_ids)
]

print("Total evaluation questions:", len(config_a_df))
print("Questions already completed:", len(completed_ids))
print("Questions remaining:", len(remaining_questions))

# Process each remaining question
for index, row in remaining_questions.iterrows():

    # Extract the question
    question = row["question"]

    # Prepare the standalone LLM prompt
    prompt = f"""<|im_start|>system
You are a research assistant specialising in brain tumour research.
Answer the question clearly and accurately using your internal knowledge.
If you do not know the answer, say that you do not know.
Do not invent facts or citations.
<|im_end|>
<|im_start|>user
{question}
<|im_end|>
<|im_start|>assistant
"""

    try:
        # Start measuring generation time
        start_time = time.time()

        # Generate the answer without document retrieval
        response = llm(
            prompt,
            max_tokens=400,
            temperature=0.1,
            stop=["<|im_end|>", "<|endoftext|>"]
        )

        # Calculate response time
        response_time = time.time() - start_time

        # Extract generated answer and finish reason
        generated_answer = response["choices"][0]["text"].strip()
        finish_reason = response["choices"][0].get("finish_reason")

        # Store all relevant evaluation information
        result = {
            "question_id": row["question_id"],
            "category": row["category"],
            "question": question,
            "gold_answer": row["gold_answer"],
            "answerable": row["answerable"],
            "supporting_sources": row["supporting_sources"],
            "generated_answer": generated_answer,
            "response_time": round(response_time, 3),
            "finish_reason": finish_reason
        }

        # Add the result to the existing results
        config_a_results = pd.concat(
            [config_a_results, pd.DataFrame([result])],
            ignore_index=True
        )

        # Save after every question to protect completed results
        config_a_results.to_csv(CONFIG_A_PATH, index=False)

        # Display progress
        print(
            f"Completed {len(config_a_results)}/80: "
            f"{row['question_id']} | "
            f"{response_time:.3f} seconds"
        )

    except Exception as error:
        print(f"\nError processing {row['question_id']}: {error}")

# Display final experiment summary
print("\nSTANDALONE LLM EVALUATION SUMMARY")
print("-" * 50)
print("Successfully generated answers:", len(config_a_results))
print("Total evaluation questions:", len(config_a_df))
print("Questions not completed:", 80 - len(config_a_results))
print("Results saved to:", CONFIG_A_PATH.resolve())
```

### 2.8 Answer Quality Evaluation Criteria

The generated answers are evaluated using a consistent scoring framework to measure their quality against the manually verified reference answers and supporting research evidence.

The following criteria are used:

| Evaluation Metric | Scoring Criteria |
|---|---|
| Correctness | 0 = Incorrect, 1 = Partially correct, 2 = Correct |
| Relevance | 0 = Irrelevant, 1 = Partially relevant, 2 = Relevant |
| Faithfulness | 0 = Unsupported or fabricated claims, 1 = Partially supported, 2 = Fully supported |
| Citation Validity | Valid, Invalid, or Not Applicable |
| Abstention | 1 = Correctly abstained, 0 = Failed to abstain, Not Applicable = Answerable question |

Correctness measures whether the generated answer agrees with the verified reference answer. Relevance measures whether the answer addresses the question. Faithfulness measures whether its claims are supported by the research evidence. Citation validity checks whether provided citations are genuine and relevant. Abstention measures whether the model appropriately avoids answering questions that cannot be answered using the available research collection.

The same evaluation criteria will be applied to all four system configurations.

### 2.9.10 Overall Performance Evaluation of Configuration A

The overall performance of Configuration A is evaluated using the manually assigned correctness, relevance and faithfulness scores. Abstention accuracy is calculated separately using the 20 unanswerable questions. Response time and generation finish reasons are also examined to assess the efficiency and limitations of the standalone LLM.

```python
import pandas as pd

# Make a copy for metric calculations
results = config_a_evaluation.copy()

# Convert evaluation scores to numeric values
score_columns = [
    "correctness_score",
    "relevance_score",
    "faithfulness_score"
]

for column in score_columns:
    results[column] = pd.to_numeric(results[column])

# Separate answerable and unanswerable questions
answerable_results = results[results["answerable"] == True]
unanswerable_results = results[results["answerable"] == False]

# Calculate normalized evaluation scores
performance_metrics = {}

for column in score_columns:
    mean_score = results[column].mean()
    performance_metrics[column] = (mean_score / 2) * 100

# Calculate exact correctness rate (score = 2)
exact_correctness = (
    results["correctness_score"] == 2
).mean() * 100

# Calculate abstention accuracy
abstention_accuracy = (
    pd.to_numeric(unanswerable_results["abstention_correct"]).mean()
) * 100

# Calculate average response time
average_response_time = results["response_time"].mean()
median_response_time = results["response_time"].median()
minimum_response_time = results["response_time"].min()
maximum_response_time = results["response_time"].max()

# Display overall performance
print("=" * 65)
print("CONFIGURATION A - OVERALL PERFORMANCE")
print("=" * 65)

print("\nEvaluation Scores (Normalized):")

for metric, score in performance_metrics.items():
    print(f"{metric}: {score:.2f}%")

print(f"\nExact Correctness Rate: {exact_correctness:.2f}%")
print(f"Abstention Accuracy: {abstention_accuracy:.2f}%")

print("\nResponse Time (seconds):")
print(f"Average: {average_response_time:.3f}")
print(f"Median: {median_response_time:.3f}")
print(f"Minimum: {minimum_response_time:.3f}")
print(f"Maximum: {maximum_response_time:.3f}")

print("\nGeneration Finish Reasons:")
print(results["finish_reason"].value_counts())

print("\n" + "=" * 65)
```

## Step 3: BM25 Index Construction

In this step, we construct a BM25 retrieval index using the
9,793 research-paper passages.

First, we tokenize and normalize the passage text by converting
it to lowercase and extracting individual words.

Next, we use the BM25Okapi algorithm from the rank_bm25
library to build the retrieval index.

The index will allow us to retrieve and rank relevant passages
for each evaluation question.

All original passages and their source metadata are preserved.

```python
# Step 3: Build the BM25 retrieval index

import re
import time
import numpy as np
from rank_bm25 import BM25Okapi

# Start measuring index construction time
start_time = time.time()

# Define a text tokenization function
def tokenize_text_eval(text):
    """
    Convert text to lowercase and extract individual words.
    """
    return re.findall(r"\b\w+\b", str(text).lower())


# Check that the retrieval dataset contains passages
if bm25_retrieval_df.empty:
    raise ValueError("The BM25 retrieval dataset is empty.")

# Tokenize all research-paper passages
bm25_corpus_eval = [
    tokenize_text_eval(text)
    for text in bm25_retrieval_df["text"]
]

# Check that all passages were processed
if len(bm25_corpus_eval) != len(bm25_retrieval_df):
    raise ValueError("Mismatch between passages and tokenized corpus.")

# Check for passages with no tokens
empty_passages = sum(
    1 for tokens in bm25_corpus_eval if len(tokens) == 0
)

if empty_passages > 0:
    raise ValueError(
        f"Found {empty_passages} passages without searchable words."
    )

# Build the BM25 index
bm25_index_eval = BM25Okapi(bm25_corpus_eval)

# Calculate construction time
elapsed_time = time.time() - start_time

# Display verification results
print("BM25 INDEX CONSTRUCTION")
print("-" * 50)

print("Total research-paper passages:", len(bm25_corpus_eval))
print("Total tokenized passages:", len(bm25_corpus_eval))
print("Empty passages:", empty_passages)
print("BM25 index created:", bm25_index_eval is not None)
print(f"Index construction time: {elapsed_time:.2f} seconds")

print("\nBM25 index is ready for retrieval testing.")
```

## Step 4: BM25 Retrieval Function

In this step, we create a search function that retrieves the
most relevant research-paper passages for a given question.

The function:
1. Tokenizes the input question.
2. Calculates BM25 relevance scores.
3. Ranks passages from highest to lowest score.
4. Returns the top-k passages.
5. Preserves article IDs, titles, page numbers, and chunk IDs
   for citation and source verification.

The retrieved passages will later be provided to the Qwen
language model to generate evidence-based answers.

```python
# Step 4: Create the BM25 search function

def bm25_search_eval(query, top_k=5):

    # Check that the query is not empty
    if not isinstance(query, str) or not query.strip():
        raise ValueError("Please provide a non-empty search query.")

    # Check that top_k is a positive integer
    if not isinstance(top_k, int) or top_k <= 0:
        raise ValueError("top_k must be a positive integer.")

    # Tokenize the user's question
    query_tokens = tokenize_text_eval(query)

    # Calculate BM25 relevance scores
    scores = bm25_index_eval.get_scores(query_tokens)

    # Retrieve the indices of the highest-scoring passages
    top_indices = np.argsort(scores)[::-1][:top_k]

    # Retrieve the corresponding passages
    results = bm25_retrieval_df.iloc[top_indices].copy()

    # Add retrieval scores and ranking positions
    results["retrieval_score"] = scores[top_indices]
    results["rank"] = range(1, len(results) + 1)

    return results


print("BM25 search function created successfully.")
```

## Step 5: BM25 Retrieval Testing

Before generating answers, we evaluate whether BM25 retrieves
relevant research-paper passages for the evaluation questions.

We begin with three factual questions (Q001-Q003).

For each question, we retrieve the top 5 passages and inspect:
- Retrieved article ID.
- Research-paper title.
- Page number.
- BM25 relevance score.
- Retrieved passage text.

We compare the retrieved sources with the expected supporting
sources in the evaluation dataset.

This pilot test helps identify retrieval problems before
generating answers for all 80 questions.

```python
# Step 5: Test BM25 retrieval using Q001-Q003

# Select the first three factual questions
test_questions = evaluation_df[
    evaluation_df["question_id"].isin(["Q001", "Q002", "Q003"])
]

# Retrieve and display the top 5 passages for each question
for _, question_row in test_questions.iterrows():

    question_id = question_row["question_id"]
    question = question_row["question"]

    print("\n" + "=" * 80)
    print(f"QUESTION ID: {question_id}")
    print(f"QUESTION: {question}")
    print("=" * 80)

    # Perform BM25 retrieval
    retrieved_passages = bm25_search_eval(
        query=question,
        top_k=5
    )

    # Display the retrieved passages
    for _, passage in retrieved_passages.iterrows():

        print(f"\nRank: {passage['rank']}")
        print(f"Article ID: {passage['article_id']}")
        print(f"Title: {passage['title']}")
        print(f"Page: {passage['page_number']}")
        print(f"BM25 Score: {passage['retrieval_score']:.4f}")
        print(f"Chunk ID: {passage['chunk_id']}")
        print(f"Passage: {passage['text'][:700]}")
        print("-" * 80)
```

## Step 6: Expected Source Ranking Analysis

The initial BM25 retrieval test did not return the expected
supporting papers among the top five results.

Before modifying the retrieval method, we investigate whether
the expected papers are present in the corpus and where they
rank in the complete BM25 search results.

For Q001, Q002, and Q003, we will:
1. Identify the expected supporting paper from the evaluation dataset.
2. Find its article ID in the BM25 corpus.
3. Calculate its ranking across all 9,793 passages.
4. Identify the highest-ranked passage from the expected paper.
5. Check the ranking of the expected supporting page.

This diagnostic will help us understand whether the problem
is caused by missing documents or poor retrieval ranking.

```python
# Step 6: Analyze the ranking of expected supporting papers

import ast

# Select the three pilot questions
pilot_questions = evaluation_df[
    evaluation_df["question_id"].isin(["Q001", "Q002", "Q003"])
]

for _, question_row in pilot_questions.iterrows():

    question_id = question_row["question_id"]
    question = question_row["question"]

    # Convert supporting sources into Python objects
    supporting_sources = question_row["supporting_sources"]

    if isinstance(supporting_sources, str):
        supporting_sources = ast.literal_eval(supporting_sources)

    print("\n" + "=" * 80)
    print(f"QUESTION: {question_id}")
    print(f"Question: {question}")

    # Calculate BM25 scores for the complete corpus
    query_tokens = tokenize_text_eval(question)
    scores = bm25_index_eval.get_scores(query_tokens)

    # Rank every passage
    ranked_indices = np.argsort(scores)[::-1]

    # Create a mapping from passage index to rank
    passage_ranks = np.empty(len(ranked_indices), dtype=int)
    passage_ranks[ranked_indices] = np.arange(
        1, len(ranked_indices) + 1
    )

    for source in supporting_sources:

        expected_title = source["title"]
        expected_page = source["page"]

        # Find article IDs matching the expected paper title
        matching_articles = bm25_retrieval_df.loc[
            bm25_retrieval_df["title"].str.strip() == expected_title.strip(),
            "article_id"
        ].unique()

        print("\nExpected paper:", expected_title)
        print("Expected page:", expected_page)
        print("Matching article IDs:", matching_articles.tolist())

        if len(matching_articles) == 0:
            print("RESULT: Expected paper was not found in the corpus.")
            continue

        # Find all passages belonging to the expected paper
        paper_mask = bm25_retrieval_df["article_id"].isin(
            matching_articles
        )

        paper_indices = np.where(paper_mask)[0]

        # Find the highest-ranked passage from the expected paper
        best_paper_index = paper_indices[
            np.argmin(passage_ranks[paper_indices])
        ]

        print(
            "Best rank from expected paper:",
            passage_ranks[best_paper_index]
        )
        print(
            "Best passage:",
            bm25_retrieval_df.iloc[best_paper_index]["chunk_id"]
        )
        print(
            "BM25 score:",
            round(scores[best_paper_index], 4)
        )

        # Check passages from the expected page
        page_mask = (
            paper_mask
            & (
                bm25_retrieval_df["page_number"].astype(str)
                == str(expected_page)
            )
        )

        page_indices = np.where(page_mask)[0]

        if len(page_indices) > 0:

            best_page_index = page_indices[
                np.argmin(passage_ranks[page_indices])
            ]

            print(
                "Best rank on expected page:",
                passage_ranks[best_page_index]
            )
            print(
                "Expected-page passage:",
                bm25_retrieval_df.iloc[best_page_index]["chunk_id"]
            )
            print(
                "Expected-page BM25 score:",
                round(scores[best_page_index], 4)
            )

        else:
            print("No passages found on the expected page.")

print("\nExpected source ranking analysis completed.")
```

## Step 11: Compare Potential Duplicate Paper Contents

The previous investigation identified 12 pairs of records with matching
paper titles and publication metadata.

In this step, I compare their extracted passages to determine whether
the records contain identical text.

This is an inspection-only step. No records are deleted or modified.
The results will help determine whether further deduplication is necessary.

```python
from collections import Counter
import re
import pandas as pd

# Normalize text to avoid differences caused only by
# capitalization and extra whitespace.
def normalize_passage(text):
    text = str(text).lower()
    text = re.sub(r"\s+", " ", text).strip()
    return text

comparison_results = []

# Use the 12 matching-title groups already identified
for title in duplicate_titles:

    article_ids = (
        bm25_clean_df.loc[
            bm25_clean_df["title"] == title, "article_id"
        ]
        .drop_duplicates()
        .tolist()
    )

    for i in range(len(article_ids)):
        for j in range(i + 1, len(article_ids)):

            article_a = article_ids[i]
            article_b = article_ids[j]

            # Get all passages for each article
            passages_a = [
                normalize_passage(x)
                for x in bm25_clean_df.loc[
                    bm25_clean_df["article_id"] == article_a, "text"
                ].dropna()
            ]

            passages_b = [
                normalize_passage(x)
                for x in bm25_clean_df.loc[
                    bm25_clean_df["article_id"] == article_b, "text"
                ].dropna()
            ]

            # Count exact matching passages, including repeated passages
            counter_a = Counter(passages_a)
            counter_b = Counter(passages_b)

            common_count = sum(
                (counter_a & counter_b).values()
            )

            comparison_results.append({
                "article_id_A": article_a,
                "article_id_B": article_b,
                "passages_A": len(passages_a),
                "passages_B": len(passages_b),
                "exact_matches": common_count,
                "overlap_A_percent": round(
                    common_count / len(passages_a) * 100, 2
                ) if passages_a else 0,
                "overlap_B_percent": round(
                    common_count / len(passages_b) * 100, 2
                ) if passages_b else 0
            })

passage_comparison_df = pd.DataFrame(comparison_results)

print("PASSAGE CONTENT COMPARISON")
print("-" * 65)
print("Paper pairs compared:", len(passage_comparison_df))

display(passage_comparison_df)
```

this step compares the extracted passage text for each pair of article IDs.

Text is normalized for capitalization and whitespace before comparison.
The comparison is inspection-only; no records are removed or modified.

```python
from collections import Counter
import re
import pandas as pd

# Normalize passage text before comparison
def normalize_passage(text):
    text = str(text).lower()
    text = re.sub(r"\s+", " ", text).strip()
    return text

comparison_results = []

# Use the verified matching titles, not the previous variable
for title in duplicate_title_check.index:

    article_ids = (
        bm25_clean_df.loc[
            bm25_clean_df["title"] == title, "article_id"
        ]
        .drop_duplicates()
        .tolist()
    )

    if len(article_ids) != 2:
        continue

    article_a, article_b = article_ids

    passages_a = [
        normalize_passage(x)
        for x in bm25_clean_df.loc[
            bm25_clean_df["article_id"] == article_a, "text"
        ].dropna()
    ]

    passages_b = [
        normalize_passage(x)
        for x in bm25_clean_df.loc[
            bm25_clean_df["article_id"] == article_b, "text"
        ].dropna()
    ]

    # Count exact text matches, including repeated passages
    counter_a = Counter(passages_a)
    counter_b = Counter(passages_b)

    common_count = sum((counter_a & counter_b).values())

    comparison_results.append({
        "title": title,
        "article_id_A": article_a,
        "article_id_B": article_b,
        "passages_A": len(passages_a),
        "passages_B": len(passages_b),
        "exact_matches": common_count,
        "overlap_A_percent": round(
            common_count / len(passages_a) * 100, 2
        ) if passages_a else 0,
        "overlap_B_percent": round(
            common_count / len(passages_b) * 100, 2
        ) if passages_b else 0
    })

passage_comparison_df = pd.DataFrame(comparison_results)

print("VERIFIED PASSAGE CONTENT COMPARISON")
print("-" * 65)
print("Paper pairs compared:", len(passage_comparison_df))

display(passage_comparison_df)
```

## Step 13: Rebuild BM25 Index After Deduplication

After removing the 12 verified duplicate paper records, I rebuild the BM25
index using the deduplicated corpus containing 7,787 passages from 175 papers.

The original BM25 index is preserved so that retrieval performance can be
compared before and after deduplication.

The rebuilt index will be used for the subsequent Configuration B retrieval
and answer-generation process.

```python
import time
import numpy as np
from rank_bm25 import BM25Okapi

# Start measuring index construction time
start_time = time.time()

# Tokenize passages from the deduplicated corpus
bm25_deduplicated_corpus = [
    tokenize_text_eval(text)
    for text in bm25_deduplicated_df["text"]
]

# Validate the tokenized corpus
if len(bm25_deduplicated_corpus) != len(bm25_deduplicated_df):
    raise ValueError("Passage and tokenized corpus counts do not match.")

empty_passages = sum(
    1 for tokens in bm25_deduplicated_corpus if not tokens
)

if empty_passages > 0:
    raise ValueError(
        f"Found {empty_passages} empty passages."
    )

# Build a new BM25 index without overwriting the original index
bm25_index_deduplicated = BM25Okapi(
    bm25_deduplicated_corpus
)

# Calculate construction time
construction_time = time.time() - start_time

# Display verification results
print("DEDUPLICATED BM25 INDEX")
print("-" * 60)
print("Total passages:", len(bm25_deduplicated_df))
print("Tokenized passages:", len(bm25_deduplicated_corpus))
print("Empty passages:", empty_passages)
print("Unique research papers:",
      bm25_deduplicated_df["article_id"].nunique())
print("Index created:", bm25_index_deduplicated is not None)
print("Construction time:", round(construction_time, 2), "seconds")
```

## Step 14: Create BM25 Search Function

This step creates a search function using the newly rebuilt BM25 index.

The function ranks research-paper passages according to their relevance
to a user's query and returns the top-k passages with their metadata
and retrieval scores.

```python
def bm25_search_deduplicated(query, top_k=5):

    # Validate the search query
    if not isinstance(query, str) or not query.strip():
        raise ValueError("Please provide a non-empty search query.")

    if not isinstance(top_k, int) or top_k <= 0:
        raise ValueError("top_k must be a positive integer.")

    # Tokenize the query
    query_tokens = tokenize_text_eval(query)

    # Calculate BM25 relevance scores
    scores = bm25_index_deduplicated.get_scores(query_tokens)

    # Retrieve the indices of the highest-scoring passages
    top_indices = np.argsort(scores)[::-1][:top_k]

    # Retrieve passages and their metadata
    results = bm25_deduplicated_df.iloc[top_indices].copy()

    # Add retrieval scores and ranking positions
    results["retrieval_score"] = scores[top_indices]
    results["rank"] = range(1, len(results) + 1)

    return results


print("Deduplicated BM25 search function created successfully.")
```

## Step 15: Test BM25 Retrieval on Initial Questions

In this step, I test the deduplicated BM25 retrieval system using the first

For each question, the system retrieves the top five passages.
The retrieved article titles, article IDs, page numbers, and relevance
scores are inspected against the expected supporting sources.

This test helps determine whether the retrieval system can locate
the correct research papers before answer generation.

```python
# Test the first three evaluation questions
test_questions = evaluation_df.head(3)

for _, question_row in test_questions.iterrows():

    question_id = question_row["question_id"]
    question = question_row["question"]

    print("\n" + "=" * 90)
    print(f"QUESTION: {question_id}")
    print("Question:", question)

    # Retrieve the top five passages
    retrieved_results = bm25_search_deduplicated(
        question,
        top_k=5
    )

    # Display the retrieved results
    display(
        retrieved_results[
            [
                "article_id",
                "title",
                "page_number",
                "section_heading",
                "retrieval_score",
                "rank"
            ]
        ]
    )

    # Display the expected supporting sources
    print("Expected supporting sources:")
    print(question_row["supporting_sources"])
```

```python
# Retrieve the top five passages for every evaluation question.

bm25_all_results = []

for _, question_row in evaluation_df.iterrows():

    question_id = question_row["question_id"]
    original_question = question_row["question"]

    # Use selected diagnostic queries where available.
    # Otherwise, use the original evaluation question.
    retrieval_query = selected_diagnostic_queries.get(
        question_id,
        original_question
    )

    # Retrieve top five passages.
    retrieved_passages = bm25_search_deduplicated(
        retrieval_query,
        top_k=5
    )

    # Store each retrieved passage with its question ID.
    for _, passage in retrieved_passages.iterrows():

        bm25_all_results.append({
            "question_id": question_id,
            "category": question_row["category"],
            "query_used": retrieval_query,
            "rank": int(passage["rank"]),
            "chunk_id": passage["chunk_id"],
            "article_id": passage["article_id"],
            "title": passage["title"],
            "page_number": passage["page_number"],
            "section_heading": passage["section_heading"],
            "retrieval_score": float(passage["retrieval_score"]),
            "text": passage["text"]
        })

# Convert results into a DataFrame.
bm25_all_results_df = pd.DataFrame(bm25_all_results)

# Basic validation.
print("BM25 RETRIEVAL RESULTS")
print("-" * 60)
print("Evaluation questions:", len(evaluation_df))
print("Retrieved passages:", len(bm25_all_results_df))
print("Expected passages:", len(evaluation_df) * 5)
print(
    "Questions with five results:",
    (bm25_all_results_df.groupby("question_id").size() == 5).sum()
)
print(
    "Unique question IDs retrieved:",
    bm25_all_results_df["question_id"].nunique()
)

# Display a sample.
display(
    bm25_all_results_df[
        [
            "question_id",
            "rank",
            "article_id",
            "page_number",
            "section_heading",
            "retrieval_score"
        ]
    ].head(15)
)
```

## Step 22: Calculate BM25 Retrieval Metrics

In this step, I evaluate the relevance of the retrieved passages
by comparing their paper titles with the expected supporting sources
in the evaluation dataset.

Recall@5 measures how many questions retrieve at least one expected
supporting paper within the top five results.

Mean Reciprocal Rank (MRR) measures how highly the first expected
supporting paper is ranked.

Retrieval performance is also examined separately for each
question category.

```python
import ast

# Convert supporting_sources into a consistent list of dictionaries.
def parse_supporting_sources(value):

    if isinstance(value, list):
        return value

    if isinstance(value, str):
        try:
            parsed = ast.literal_eval(value)
            return parsed if isinstance(parsed, list) else []
        except (ValueError, SyntaxError):
            return []

    return []


# Store expected supporting sources for each question.
expected_sources = {}

for _, row in evaluation_df.iterrows():

    sources = parse_supporting_sources(row["supporting_sources"])

    expected_sources[row["question_id"]] = {
        str(source.get("title", "")).strip().casefold()
        for source in sources
        if isinstance(source, dict) and source.get("title")
    }


# Evaluate the first expected source found in the top five.
retrieval_evaluation = []

for question_id, group in bm25_all_results_df.groupby(
    "question_id", sort=True
):

    expected_titles = expected_sources[question_id]

    retrieved_titles = [
        str(title).strip().casefold()
        for title in group.sort_values("rank")["title"]
    ]

    # Find the rank of the first expected supporting paper.
    matching_ranks = [
        rank
        for rank, title in enumerate(retrieved_titles, start=1)
        if title in expected_titles
    ]

    first_relevant_rank = (
        min(matching_ranks) if matching_ranks else None
    )

    hit_at_5 = int(first_relevant_rank is not None)

    reciprocal_rank = (
        1 / first_relevant_rank if first_relevant_rank else 0
    )

    retrieval_evaluation.append({
        "question_id": question_id,
        "category": group["category"].iloc[0],
        "expected_source_count": len(expected_titles),
        "first_relevant_rank": first_relevant_rank,
        "hit_at_5": hit_at_5,
        "reciprocal_rank": reciprocal_rank
    })


retrieval_evaluation_df = pd.DataFrame(retrieval_evaluation)

# Overall retrieval metrics.
recall_at_5 = retrieval_evaluation_df["hit_at_5"].mean()
mrr = retrieval_evaluation_df["reciprocal_rank"].mean()

print("OVERALL BM25 RETRIEVAL PERFORMANCE")
print("-" * 60)
print(f"Recall@5: {recall_at_5:.2%}")
print(f"MRR:      {mrr:.4f}")
print(
    "Questions with an expected source in top 5:",
    int(retrieval_evaluation_df["hit_at_5"].sum()),
    "/",
    len(retrieval_evaluation_df)
)

# Performance by category.
category_metrics = (
    retrieval_evaluation_df
    .groupby("category")
    .agg(
        questions=("question_id", "count"),
        hits_at_5=("hit_at_5", "sum"),
        recall_at_5=("hit_at_5", "mean"),
        mrr=("reciprocal_rank", "mean")
    )
    .reset_index()
)

print("\nPERFORMANCE BY QUESTION CATEGORY")
print("-" * 60)
display(category_metrics)

# Inspect questions where no expected source was retrieved.
missed_questions = retrieval_evaluation_df[
    retrieval_evaluation_df["hit_at_5"] == 0
]

print("\nQUESTIONS WITHOUT AN EXPECTED SOURCE IN TOP 5")
print("-" * 60)
print("Number of missed questions:", len(missed_questions))
display(missed_questions)
```

## Step 23: Diagnose Expected Source Coverage

In this step, I investigate the 60 answerable evaluation questions.

For each question, I calculate the best rank of its expected
supporting papers across the entire deduplicated corpus.

This identifies whether a missed source is absent from the corpus
or present but ranked outside the top five.

Unanswerable questions are excluded because they have no expected
supporting papers.

```python
# Diagnose expected source coverage and ranking across the full corpus.

source_diagnostics = []

for _, row in evaluation_df.iterrows():

    question_id = row["question_id"]

    # Exclude unanswerable questions from source-coverage analysis.
    if not row["answerable"]:
        continue

    # Use the same query that was used in the initial evaluation.
    query = selected_diagnostic_queries.get(
        question_id,
        row["question"]
    )

    query_tokens = tokenize_text_eval(query)
    all_scores = bm25_index_deduplicated.get_scores(query_tokens)

    expected_titles = expected_sources[question_id]

    matched_sources = []

    for expected_title in expected_titles:

        # Find the expected paper in the corpus.
        title_mask = (
            bm25_deduplicated_df["title"]
            .astype(str)
            .str.strip()
            .str.casefold()
            .to_numpy() == expected_title
        )

        matching_indices = np.where(title_mask)[0]

        if len(matching_indices) == 0:
            matched_sources.append({
                "title": expected_title,
                "present_in_corpus": False,
                "best_rank": None,
                "best_score": None
            })
            continue

        # Find the best passage belonging to this expected paper.
        best_index = matching_indices[
            np.argmax(all_scores[matching_indices])
        ]

        best_rank = int(
            np.sum(all_scores > all_scores[best_index]) + 1
        )

        matched_sources.append({
            "title": expected_title,
            "present_in_corpus": True,
            "best_rank": best_rank,
            "best_score": round(all_scores[best_index], 4)
        })

    # Select the highest-ranked expected paper for this question.
    available_sources = [
        source for source in matched_sources
        if source["present_in_corpus"]
    ]

    best_source = (
        min(available_sources, key=lambda x: x["best_rank"])
        if available_sources else None
    )

    source_diagnostics.append({
        "question_id": question_id,
        "category": row["category"],
        "expected_source_count": len(expected_titles),
        "sources_found_in_corpus": len(available_sources),
        "best_expected_rank": (
            best_source["best_rank"] if best_source else None
        ),
        "best_expected_score": (
            best_source["best_score"] if best_source else None
        ),
        "source_missing_from_corpus": (
            len(available_sources) == 0
        ),
        "source_within_top_5": (
            best_source["best_rank"] <= 5 if best_source else False
        )
    })


source_diagnostics_df = pd.DataFrame(source_diagnostics)

# Summarise corpus coverage and ranking.
print("EXPECTED SOURCE COVERAGE")
print("-" * 60)
print("Answerable questions:", len(source_diagnostics_df))
print(
    "Questions with at least one expected paper in corpus:",
    int((~source_diagnostics_df["source_missing_from_corpus"]).sum())
)
print(
    "Questions with no expected paper in corpus:",
    int(source_diagnostics_df["source_missing_from_corpus"].sum())
)
print(
    "Questions with an expected paper ranked in top 5:",
    int(source_diagnostics_df["source_within_top_5"].sum())
)

# Display questions where expected papers exist but rank below 5.
print("\nEXPECTED SOURCES PRESENT BUT RANKED BELOW TOP 5")
print("-" * 60)

ranked_below_top5 = source_diagnostics_df[
    (~source_diagnostics_df["source_missing_from_corpus"])
    & (~source_diagnostics_df["source_within_top_5"])
]

display(
    ranked_below_top5.sort_values("best_expected_rank")
)

# Display questions whose expected papers are absent.
print("\nEXPECTED SOURCES NOT FOUND IN CORPUS")
print("-" * 60)

missing_from_corpus = source_diagnostics_df[
    source_diagnostics_df["source_missing_from_corpus"]
]

display(missing_from_corpus)
```

## Verify Missing Expected Sources

This check confirms whether any expected supporting papers
are completely absent from the deduplicated BM25 corpus.

```python
print("SOURCE COVERAGE SUMMARY")
print("-" * 50)

print(
    "Answerable questions:",
    len(source_diagnostics_df)
)

print(
    "Expected sources missing from corpus:",
    source_diagnostics_df["source_missing_from_corpus"].sum()
)

print(
    "Expected sources present in corpus:",
    (~source_diagnostics_df["source_missing_from_corpus"]).sum()
)

print("\nMISSING QUESTIONS:")
display(
    missing_from_corpus[
        ["question_id", "category", "expected_source_count"]
    ]
)

print("\nSOURCES PRESENT BUT RANKED BELOW TOP 5:")
display(
    ranked_below_top5[
        ["question_id", "category", "best_expected_rank"]
    ].sort_values("best_expected_rank")
)
```

## Step 24: Inspect Missed Retrieval Questions

In this step, I examine the 24 answerable questions whose expected
supporting papers were not retrieved in the top five.

I display each question, its expected paper titles, its best
retrieval rank, and the query used.

This analysis helps identify retrieval weaknesses before selecting
an improvement strategy.

```python
# Inspect questions whose expected sources were ranked below top 5.

missed_ids = set(
    ranked_below_top5["question_id"]
)

missed_question_details = []

for _, row in evaluation_df.iterrows():

    question_id = row["question_id"]

    if question_id not in missed_ids:
        continue

    sources = parse_supporting_sources(
        row["supporting_sources"]
    )

    expected_titles = [
        source.get("title", "")
        for source in sources
        if isinstance(source, dict)
    ]

    retrieval_query = selected_diagnostic_queries.get(
        question_id,
        row["question"]
    )

    diagnostic_row = source_diagnostics_df[
        source_diagnostics_df["question_id"] == question_id
    ].iloc[0]

    missed_question_details.append({
        "question_id": question_id,
        "category": row["category"],
        "question": row["question"],
        "expected_papers": " | ".join(expected_titles),
        "best_expected_rank": diagnostic_row["best_expected_rank"],
        "query_used": retrieval_query
    })


missed_question_details_df = pd.DataFrame(
    missed_question_details
)

print("MISSED RETRIEVAL QUESTIONS")
print("-" * 70)
print("Total questions:", len(missed_question_details_df))

display(
    missed_question_details_df.sort_values(
        ["category", "best_expected_rank"]
    ).reset_index(drop=True)
)
```

## Step 25: Verify Individual Expected Paper Coverage

Some evaluation questions require evidence from multiple research papers.

This step checks every expected supporting paper individually,
rather than checking only whether at least one expected paper
is available for each question.

This helps distinguish missing corpus records from retrieval
ranking problems, especially for multi-paper comparison questions.

```python
# Create a set of all paper titles available in the BM25 corpus.
corpus_titles = set(
    bm25_deduplicated_df["title"]
    .astype(str)
    .str.strip()
    .str.casefold()
)

paper_coverage_records = []

for question_id, expected_titles in expected_sources.items():

    for expected_title in expected_titles:

        paper_coverage_records.append({
            "question_id": question_id,
            "expected_title": expected_title,
            "present_in_corpus": expected_title in corpus_titles
        })

paper_coverage_df = pd.DataFrame(paper_coverage_records)

# Overall paper-level coverage.
total_expected_papers = len(paper_coverage_df)

present_papers = int(
    paper_coverage_df["present_in_corpus"].sum()
)

missing_papers = paper_coverage_df[
    ~paper_coverage_df["present_in_corpus"]
]

print("INDIVIDUAL EXPECTED PAPER COVERAGE")
print("-" * 60)
print("Expected paper references:", total_expected_papers)
print("Present in corpus:", present_papers)
print("Missing from corpus:", len(missing_papers))

if total_expected_papers > 0:
    print(
        "Paper-level coverage:",
        f"{present_papers / total_expected_papers:.2%}"
    )

print("\nMISSING EXPECTED PAPERS")
print("-" * 60)
display(missing_papers)
```

## Step 26: Investigate Missing Paper Records

Two expected paper references are absent from the deduplicated
BM25 corpus.

In this step, I compare their titles against the original BM25
dataset and the active article metadata.

I also identify similar titles to check whether the papers
were recorded under alternative title formats.

No corpus records are modified in this step.

```python
from difflib import SequenceMatcher

# Normalize titles for comparison.
def normalize_title(title):
    return " ".join(
        str(title).casefold().strip().split()
    )


# Find the title column in the active metadata.
possible_title_columns = [
    column for column in active_check_df.columns
    if column.strip().casefold() in ["title", "article_title"]
]

if not possible_title_columns:
    raise ValueError(
        "Could not identify the title column in active_check_df. "
        f"Available columns: {list(active_check_df.columns)}"
    )

active_title_column = possible_title_columns[0]

# Collect titles from each available dataset.
original_bm25_titles = (
    bm25_retrieval_df["title"].dropna().unique().tolist()
)

clean_bm25_titles = (
    bm25_clean_df["title"].dropna().unique().tolist()
)

active_titles = (
    active_check_df[active_title_column].dropna().unique().tolist()
)

comparison_datasets = {
    "Original BM25 dataset": original_bm25_titles,
    "Reference-cleaned BM25 dataset": clean_bm25_titles,
    "Active metadata": active_titles
}

# Compare the two missing expected titles.
missing_title_records = []

for _, missing_row in missing_papers.iterrows():

    expected_title = missing_row["expected_title"]
    normalized_expected = normalize_title(expected_title)

    print("\n" + "=" * 75)
    print("QUESTION:", missing_row["question_id"])
    print("EXPECTED TITLE:", expected_title)

    for dataset_name, titles in comparison_datasets.items():

        normalized_titles = [
            (title, normalize_title(title))
            for title in titles
        ]

        exact_matches = [
            original
            for original, normalized in normalized_titles
            if normalized == normalized_expected
        ]

        similar_matches = sorted(
            [
                (
                    original,
                    SequenceMatcher(
                        None,
                        normalized_expected,
                        normalized
                    ).ratio()
                )
                for original, normalized in normalized_titles
            ],
            key=lambda item: item[1],
            reverse=True
        )[:3]

        print(f"\n{dataset_name}")
        print("Exact matches:", exact_matches)

        print("Closest title matches:")
        for title, similarity in similar_matches:
            print(f"  Similarity {similarity:.3f}: {title}")
```

## Step 27: Verify Two Paper Records

The previous inspection identified two likely title-formatting
problems rather than two missing research papers.

This step locates the corresponding records in the BM25 corpus,
checks their article IDs, and counts their extracted passages.

No original titles or corpus records are modified.

```python
# Locate the two papers using their distinctive title fragments.

title_checks = {
    "Q037": "explainable hybrid cnn",
    "Q057": "tumor identification and classification of mri"
}

for question_id, fragment in title_checks.items():

    matches = bm25_deduplicated_df[
        bm25_deduplicated_df["title"]
        .astype(str)
        .str.casefold()
        .str.contains(fragment, regex=False)
    ]

    print("\n" + "=" * 65)
    print("QUESTION:", question_id)

    if matches.empty:
        print("No matching record found.")
        continue

    paper_summary = (
        matches.groupby(
            ["article_id", "title"],
            as_index=False
        )
        .agg(
            passage_count=("chunk_id", "count"),
            first_page=("page_number", "min"),
            last_page=("page_number", "max")
        )
    )

    display(paper_summary)
```

## Step 28: Correct Title Matching for Evaluation

The investigation confirmed that two expected papers are present
in the corpus but have title-formatting differences.

Q037 contains an encoding error in an en dash.
Q057 is missing the first letter of its title.

In this step, I create explicit title aliases for these two records.
The original corpus, evaluation questions, and gold answers remain
unchanged.

I then recalculate paper-level coverage, Recall@5, and MRR using
the corrected title matching.

```python
# Explicit title aliases based on the verified article records.
# These aliases are used only for evaluation matching.

expected_title_aliases = {
    normalize_title(
        "An explainable hybrid CNN–vision transformer framework "
        "for brain tumor detection and classification using MRI"
    ): normalize_title(
        "An explainable hybrid CNNâ€“vision transformer framework "
        "for brain tumor detection and classification using MRI"
    ),

    normalize_title(
        "Brain tumor identification and classification of MRI images "
        "using deep learning techniques"
    ): normalize_title(
        "rain Tumor Identification and Classification of MRI Images "
        "Using Deep Learning Techniques"
    )
}


# Create corrected expected source titles for every question.
corrected_expected_sources = {}

for question_id, titles in expected_sources.items():

    corrected_expected_sources[question_id] = {
        expected_title_aliases.get(title, normalize_title(title))
        for title in titles
    }


# Check all individual expected paper references.
corpus_titles = set(
    bm25_deduplicated_df["title"]
    .dropna()
    .apply(normalize_title)
)

coverage_records = []

for question_id, titles in corrected_expected_sources.items():

    for title in titles:

        coverage_records.append({
            "question_id": question_id,
            "expected_title": title,
            "present_in_corpus": title in corpus_titles
        })

corrected_coverage_df = pd.DataFrame(coverage_records)


# Recalculate Recall@5 and MRR.
corrected_retrieval_records = []

for question_id, group in bm25_all_results_df.groupby(
    "question_id", sort=True
):

    expected_titles = corrected_expected_sources[question_id]

    retrieved_titles = (
        group.sort_values("rank")["title"]
        .apply(normalize_title)
        .tolist()
    )

    matching_ranks = [
        rank
        for rank, title in enumerate(retrieved_titles, start=1)
        if title in expected_titles
    ]

    first_rank = min(matching_ranks) if matching_ranks else None

    corrected_retrieval_records.append({
        "question_id": question_id,
        "category": group["category"].iloc[0],
        "first_relevant_rank": first_rank,
        "hit_at_5": int(first_rank is not None),
        "reciprocal_rank": (
            1 / first_rank if first_rank is not None else 0
        )
    })


corrected_retrieval_df = pd.DataFrame(
    corrected_retrieval_records
)


# Report corrected results.
print("CORRECTED EXPECTED PAPER COVERAGE")
print("-" * 60)
print("Expected paper references:", len(corrected_coverage_df))
print(
    "Present in corpus:",
    int(corrected_coverage_df["present_in_corpus"].sum())
)
print(
    "Missing from corpus:",
    int((~corrected_coverage_df["present_in_corpus"]).sum())
)

print("\nCORRECTED BM25 RETRIEVAL PERFORMANCE")
print("-" * 60)

print(
    "Overall Recall@5:",
    f"{corrected_retrieval_df['hit_at_5'].mean():.2%}"
)

print(
    "Answerable Recall@5:",
    f"{corrected_retrieval_df[
        corrected_retrieval_df['question_id'].isin(
            evaluation_df.loc[
                evaluation_df['answerable'],
                'question_id'
            ]
        )
    ]['hit_at_5'].mean():.2%}"
)

print(
    "MRR:",
    f"{corrected_retrieval_df['reciprocal_rank'].mean():.4f}"
)

print(
    "Questions with expected source in top 5:",
    int(corrected_retrieval_df["hit_at_5"].sum()),
    "/ 80"
)

print("\nQUESTIONS STILL MISSING FROM TOP 5")
print("-" * 60)

display(
    corrected_retrieval_df[
        corrected_retrieval_df["hit_at_5"] == 0
    ].sort_values("question_id")
)
```

## Step 29: Evaluate Original BM25 Queries

In this step, I evaluate BM25 using the original 80 evaluation
questions without manually expanded queries.

This provides a fair baseline for Configuration B.

The corrected expected-paper title aliases are retained for
evaluation matching only. The original questions and gold answers
remain unchanged.

The results will be used to compare BM25 against subsequent
retrieval improvements.

```python
# Evaluate all 80 questions using their original wording.
# No manually expanded queries are used.

bm25_original_results = []

for _, row in evaluation_df.iterrows():

    question_id = row["question_id"]
    original_query = row["question"]

    results = bm25_search_deduplicated(
        original_query,
        top_k=5
    )

    for _, passage in results.iterrows():

        bm25_original_results.append({
            "question_id": question_id,
            "category": row["category"],
            "rank": int(passage["rank"]),
            "article_id": passage["article_id"],
            "title": passage["title"],
            "chunk_id": passage["chunk_id"],
            "page_number": passage["page_number"],
            "retrieval_score": float(passage["retrieval_score"]),
            "text": passage["text"]
        })


bm25_original_results_df = pd.DataFrame(
    bm25_original_results
)


# Evaluate expected-paper matching.
original_evaluation_records = []

for question_id, group in bm25_original_results_df.groupby(
    "question_id", sort=True
):

    expected_titles = corrected_expected_sources[question_id]

    retrieved_titles = (
        group.sort_values("rank")["title"]
        .apply(normalize_title)
        .tolist()
    )

    matching_ranks = [
        rank
        for rank, title in enumerate(retrieved_titles, start=1)
        if title in expected_titles
    ]

    first_rank = min(matching_ranks) if matching_ranks else None

    original_evaluation_records.append({
        "question_id": question_id,
        "category": group["category"].iloc[0],
        "first_relevant_rank": first_rank,
        "hit_at_5": int(first_rank is not None),
        "reciprocal_rank": (
            1 / first_rank if first_rank is not None else 0
        )
    })


bm25_original_evaluation_df = pd.DataFrame(
    original_evaluation_records
)


# Overall metrics.
overall_hit_rate = (
    bm25_original_evaluation_df["hit_at_5"].mean()
)

overall_mrr = (
    bm25_original_evaluation_df["reciprocal_rank"].mean()
)

answerable_ids = set(
    evaluation_df.loc[
        evaluation_df["answerable"],
        "question_id"
    ]
)

answerable_results = bm25_original_evaluation_df[
    bm25_original_evaluation_df["question_id"].isin(answerable_ids)
]

answerable_hit_rate = answerable_results["hit_at_5"].mean()


print("ORIGINAL BM25 BASELINE")
print("-" * 60)
print("Questions:", len(evaluation_df))
print("Retrieved passages:", len(bm25_original_results_df))
print(f"Overall hit rate@5: {overall_hit_rate:.2%}")
print(f"Answerable hit rate@5: {answerable_hit_rate:.2%}")
print(f"MRR: {overall_mrr:.4f}")

# Category-level performance.
original_category_metrics = (
    bm25_original_evaluation_df
    .groupby("category")
    .agg(
        questions=("question_id", "count"),
        hits_at_5=("hit_at_5", "sum"),
        hit_rate_at_5=("hit_at_5", "mean"),
        mrr=("reciprocal_rank", "mean")
    )
    .reset_index()
)

print("\nPERFORMANCE BY CATEGORY")
display(original_category_metrics)
```

## Step 31: Prepare Labelled Evidence for Citation Testing

In this step, I prepare the five passages retrieved by BM25

Each passage receives a temporary source label, such as [S1]
or [S2]. These labels allow the language model to identify
which retrieved passages support its answer.

The original article metadata is preserved so that each
source label can later be converted into a traceable citation.

```python
# Prepare the top five BM25 passages for Q001.
# Each passage receives a temporary source label.

q001_results = (
    bm25_original_results_df[
        bm25_original_results_df["question_id"] == "Q001"
    ]
    .sort_values("rank")
    .copy()
)

citation_source_map = {}

context_parts = []

for _, row in q001_results.iterrows():

    # Create a temporary source label.
    source_label = f"S{int(row['rank'])}"

    # Preserve the actual citation metadata.
    citation_source_map[source_label] = {
        "article_id": row["article_id"],
        "title": row["title"],
        "page_number": row["page_number"],
        "chunk_id": row["chunk_id"]
    }

    # Add the labelled passage to the evidence context.
    context_parts.append(
        f"[{source_label}]\n"
        f"Article: {row['title']}\n"
        f"Page: {row['page_number']}\n"
        f"Passage:\n{row['text']}"
    )

q001_context = "\n\n".join(context_parts)

print("QUESTION:")
print(
    evaluation_df.loc[
        evaluation_df["question_id"] == "Q001",
        "question"
    ].iloc[0]
)

print("\nSOURCE LABELS AND CITATION METADATA:")
for label, metadata in citation_source_map.items():
    print(f"\n[{label}]")
    print(metadata)

print("\nTOTAL EVIDENCE CHARACTERS:", len(q001_context))
```

## Step 38: Test BM25 Evidence Selection

The initial BM25 results included passages that mentioned
brain tumour detection but did not directly answer the question.

I test a transparent keyword-based evidence-selection method
on the top ten retrieved passages.

The method scores passages using relevant machine learning
and brain tumour detection terminology.

This is a diagnostic test before deciding the final
evidence-selection procedure.

```python
# Score the top ten Q001 passages using relevant terminology.
# This is a diagnostic selection method, not a trained model.

import re

method_terms = [
    "machine learning",
    "deep learning",
    "convolutional neural network",
    "cnn",
    "cnn-lstm",
    "support vector machine",
    "svm",
    "decision tree",
    "random forest",
    "knn",
    "k-nearest neighbors",
    "k-means",
    "dnn",
    "vision transformer",
    "classification",
    "segmentation"
]

question_terms = [
    "brain tumor",
    "brain tumour",
    "detection",
    "classification"
]

q001_candidates = (
    bm25_search_deduplicated(question, top_k=10)
    .copy()
)

def calculate_evidence_score(text):

    text_lower = str(text).lower()

    method_score = sum(
        1 for term in method_terms
        if re.search(
            r"\b" + re.escape(term) + r"\b",
            text_lower
        )
    )

    question_score = sum(
        1 for term in question_terms
        if term in text_lower
    )

    return method_score + question_score


q001_candidates["evidence_score"] = (
    q001_candidates["text"].apply(calculate_evidence_score)
)

q001_candidates = q001_candidates.sort_values(
    "evidence_score",
    ascending=False
)

display(
    q001_candidates[
        [
            "article_id",
            "title",
            "page_number",
            "chunk_id",
            "evidence_score"
        ]
    ]
)
```

## Step 40A: Prepare Compact Evidence Context

The initial prompt exceeded the language model's context window.

To address this, I reduce the evidence context from four
passages to the three highest-scoring passages.

The selected passages are retained with their original
citation metadata to preserve traceability.

```python
# Select the three highest-scoring evidence passages.
q001_compact_evidence = (
    q001_selected_evidence
    .head(3)
    .copy()
    .reset_index(drop=True)
)

# Rebuild citation labels for the selected passages.
citation_source_map = {}
context_parts = []

for position, row in q001_compact_evidence.iterrows():

    source_label = f"S{position + 1}"

    citation_source_map[source_label] = {
        "article_id": row["article_id"],
        "title": row["title"],
        "page_number": row["page_number"],
        "chunk_id": row["chunk_id"]
    }

    context_parts.append(
        f"[{source_label}] "
        f"Title: {row['title']}\n"
        f"Page: {row['page_number']}\n"
        f"Evidence: {row['text']}"
    )

q001_selected_context = "\n\n".join(context_parts)

# Rebuild a shorter prompt.
improved_system_prompt = """
You are an academic research assistant.
Answer using only the supplied evidence.
Give a concise, direct answer.
Do not invent information.
Cite supporting sources using labels such as [S1].
Use only the supplied labels.
If the evidence cannot answer the question, write:
INSUFFICIENT_EVIDENCE.
"""

improved_user_prompt = f"""
Question: {question}

Evidence:
{q001_selected_context}

Give the answer with citations.
"""

# Check the prompt size before generating.
prompt_token_count = len(
    llm.tokenize(
        (
            improved_system_prompt
            + "\n"
            + improved_user_prompt
        ).encode("utf-8")
    )
)

print("Prompt tokens:", prompt_token_count)
print("Model context limit: 2048")
print("Remaining context tokens:", 2048 - prompt_token_count)

print("\nSelected evidence:")
for label, metadata in citation_source_map.items():
    print(label, "->", metadata)
```

## Step 40B: Generate Q001 Answer with Citations

The compact evidence context fits within the model's
available context window.

I use the Qwen language model to generate an answer
grounded in the three selected passages.

Each supporting claim should include a source label
that can be mapped to its original paper and page.

```python
import time

# Record the generation start time.
generation_start = time.perf_counter()

# Generate the answer using the compact evidence context.
improved_generation_result = llm.create_chat_completion(
    messages=[
        {
            "role": "system",
            "content": improved_system_prompt
        },
        {
            "role": "user",
            "content": improved_user_prompt
        }
    ],
    temperature=0.1,
    max_tokens=250
)

# Calculate response time.
q001_generation_time = (
    time.perf_counter() - generation_start
)

# Extract the generated answer and finish reason.
q001_improved_answer = (
    improved_generation_result["choices"][0]
    ["message"]["content"].strip()
)

q001_improved_finish_reason = (
    improved_generation_result["choices"][0]["finish_reason"]
)

# Display results.
print("QUESTION:", question)
print("\nGENERATED ANSWER:")
print(q001_improved_answer)

print("\nFINISH REASON:", q001_improved_finish_reason)
print("RESPONSE TIME:", round(q001_generation_time, 3), "seconds")

print("\nCITATION SOURCE MAP:")
for label, metadata in citation_source_map.items():
    print(label, "->", metadata)
```

## Step 40C: Extract Focused Evidence for Q001

The previous generation attempt produced an incorrect
abstention despite relevant information being present.

To investigate this issue, I extract sentences that
explicitly mention machine learning methods from the
selected retrieved passages.

This reduces unnecessary context and makes the relevant
evidence easier for the language model to identify.

```python
import re
import time

# Terms relevant to the Q001 question.
method_terms = [
    "cnn-lstm",
    "convolutional neural network",
    "cnn",
    "support vector machine",
    "svm",
    "decision tree",
    "random forest",
    "k-nearest neighbor",
    "k-nearest neighbour",
    "knn",
    "k-means",
    "dnn",
    "deep neural network"
]

focused_source_map = {}
focused_evidence_parts = []

# Extract relevant sentences from the three selected passages.
for position, (_, row) in enumerate(
    q001_compact_evidence.iterrows(),
    start=1
):
    label = f"S{position}"

    sentences = re.split(
        r"(?<=[.!?])\s+",
        str(row["text"])
    )

    relevant_sentences = []

    for sentence in sentences:
        sentence_lower = sentence.lower()

        if any(
            term in sentence_lower
            for term in method_terms
        ):
            relevant_sentences.append(sentence.strip())

    focused_source_map[label] = {
        "article_id": row["article_id"],
        "title": row["title"],
        "page_number": row["page_number"],
        "chunk_id": row["chunk_id"]
    }

    focused_evidence_parts.append(
        f"[{label}] {row['title']}, page {row['page_number']}:\n"
        + " ".join(relevant_sentences)
    )

focused_context = "\n\n".join(focused_evidence_parts)

# Display the extracted evidence before generation.
print("FOCUSED EVIDENCE")
print("=" * 70)
print(focused_context)

print("\nSOURCE METADATA")
for label, metadata in focused_source_map.items():
    print(label, "->", metadata)
```

## Step 40D: Generate Answer from Focused Evidence

The focused evidence contains explicit descriptions of
machine learning and deep learning methods.

I use this reduced evidence context to generate a concise
answer with source labels.

The generated answer is checked against the original

```python
# Use a simple extraction-focused prompt.
focused_system_prompt = """
You are an academic research assistant.

Answer the question using the evidence provided.

The evidence explicitly contains machine learning methods.
Extract and organise those methods into a concise answer.

Do not say there is insufficient evidence when methods
are explicitly listed in the passages.

Use these categories where appropriate:
1. Traditional machine learning
2. Deep learning
3. Clustering methods

Cite each group using the supplied source labels.
Do not invent methods or citations.

Return only the answer.
"""

focused_user_prompt = f"""
Question:
{question}

Evidence:
{focused_context}

Answer with source labels such as [S1] and [S2].
"""

# Generate the answer.
focused_generation_start = time.perf_counter()

focused_generation_result = llm.create_chat_completion(
    messages=[
        {
            "role": "system",
            "content": focused_system_prompt
        },
        {
            "role": "user",
            "content": focused_user_prompt
        }
    ],
    temperature=0.1,
    max_tokens=200
)

focused_generation_time = (
    time.perf_counter() - focused_generation_start
)

focused_q001_answer = (
    focused_generation_result["choices"][0]
    ["message"]["content"].strip()
)

focused_q001_finish_reason = (
    focused_generation_result["choices"][0]["finish_reason"]
)

print("GENERATED Q001 ANSWER")
print("-" * 60)
print(focused_q001_answer)

print("\nFINISH REASON:", focused_q001_finish_reason)
print(
    "RESPONSE TIME:",
    round(focused_generation_time, 3),
    "seconds"
)
```

## Step 40E: Test Direct Method Extraction

## Configuration B: Generalised Evidence Selection

In this step, I develop a reusable evidence-selection
function for BM25 retrieval results.

The function combines the original BM25 ranking with
question-term overlap to prioritise passages that contain
information related to the user's question.

It also limits the number of selected passages and their
combined text length to keep the evidence within the
language model's context window.

Article IDs, titles, page numbers, chunk IDs, and retrieval
scores are preserved to support source attribution.

The same function will be applied to factual, conceptual,
multi-paper comparison, and unanswerable questions.

```python
import re
import pandas as pd

# Terms that commonly contribute little to question matching
STOPWORDS = {
    "what", "which", "who", "whom", "where", "when",
    "why", "how", "does", "did", "do", "is", "are",
    "was", "were", "the", "a", "an", "and", "or",
    "of", "in", "on", "for", "to", "from", "with",
    "by", "their", "they", "them", "this", "that",
    "these", "those", "using", "used", "use",
    "according", "collection"
}


def select_bm25_evidence(
    question,
    retrieval_results,
    max_passages=5,
    max_characters=6500
):
    """
    Select relevant BM25 passages using question-term overlap
    and the original BM25 retrieval ranking.

    Original source metadata is preserved.
    """

    if retrieval_results is None or retrieval_results.empty:
        return pd.DataFrame()

    selected = []
    used_characters = 0

    # Extract meaningful question terms
    question_terms = {
        term.lower()
        for term in re.findall(r"\b[a-zA-Z0-9]+\b", question)
        if len(term) > 2 and term.lower() not in STOPWORDS
    }

    # Copy results so the original retrieval data is unchanged
    candidates = retrieval_results.copy()

    # Calculate question-term overlap
    def calculate_overlap(text):
        passage_terms = {
            term.lower()
            for term in re.findall(r"\b[a-zA-Z0-9]+\b", str(text))
        }

        if not question_terms:
            return 0

        return len(question_terms.intersection(passage_terms))

    candidates["question_overlap"] = candidates["text"].apply(
        calculate_overlap
    )

    # Preserve BM25 ranking as the primary signal.
    # Use overlap to help distinguish passages with similar scores.
    candidates = candidates.sort_values(
        by=["retrieval_score", "question_overlap"],
        ascending=[False, False]
    )

    for _, row in candidates.iterrows():

        passage = str(row["text"]).strip()

        if not passage:
            continue

        remaining = max_characters - used_characters

        if remaining <= 0:
            break

        # Keep the full passage if it fits; otherwise truncate it
        selected_text = passage[:remaining]

        selected_row = row.copy()
        selected_row["text"] = selected_text

        selected.append(selected_row)
        used_characters += len(selected_text)

        if len(selected) >= max_passages:
            break

    if not selected:
        return pd.DataFrame()

    return pd.DataFrame(selected).reset_index(drop=True)


print("Evidence-selection function created successfully.")
```

## Configuration B: Reusable LLM Answer Generation

In this step, I develop a reusable answer-generation
function using the Qwen2.5-0.5B-Instruct language model.

The function receives a research question and its selected
BM25 passages. Each passage is assigned a source label
linked to its article ID, title, and page number.

The language model generates an answer using only the
provided evidence. It is instructed to avoid unsupported
claims and abstain when the evidence is insufficient.

The same generation function and prompt structure are
applied consistently across all four question categories.

This maintains a consistent answer-generation procedure
for Configuration B.

```python
import time


def generate_bm25_answer(
    question,
    selected_evidence,
    max_tokens=220
):
    """
    Generate an evidence-grounded answer using Qwen
    and selected BM25 passages.

    Returns the generated answer, source mapping,
    generation time, and finish reason.
    """

    if selected_evidence is None or selected_evidence.empty:
        return {
            "answer": "Insufficient evidence in the retrieved passages.",
            "sources": {},
            "generation_time": 0,
            "finish_reason": "no_evidence"
        }

    # Construct labelled evidence and citation mapping
    evidence_blocks = []
    source_mapping = {}

    for index, (_, row) in enumerate(
        selected_evidence.iterrows(),
        start=1
    ):
        source_label = f"S{index}"

        source_mapping[source_label] = {
            "article_id": row["article_id"],
            "title": row["title"],
            "page_number": row["page_number"],
            "chunk_id": row["chunk_id"]
        }

        evidence_blocks.append(
            f"[{source_label}]\n"
            f"Article: {row['title']}\n"
            f"Page: {row['page_number']}\n"
            f"Passage: {row['text']}"
        )

    evidence_text = "\n\n".join(evidence_blocks)

    # Use one consistent prompt for every question
    prompt = f"""
You are an academic research assistant.

Answer the question using only the supplied research
evidence.

Rules:
1. Give a direct and relevant answer.
2. Include important information from the evidence.
3. Do not invent facts, numbers, or explanations.
4. Cite supporting passages using [S1], [S2], etc.
5. If the evidence does not answer the question,
   state: Insufficient evidence in the provided collection.
6. Do not guess when information is missing.

QUESTION:
{question}

RESEARCH EVIDENCE:
{evidence_text}

ANSWER:
"""

    # Generate answer
    start_time = time.time()

    response = llm(
        prompt,
        max_tokens=max_tokens,
        temperature=0.1,
        top_p=0.9,
        stop=["</s>"]
    )

    generation_time = time.time() - start_time

    answer = response["choices"][0]["text"].strip()
    finish_reason = response["choices"][0]["finish_reason"]

    return {
        "answer": answer,
        "sources": source_mapping,
        "generation_time": generation_time,
        "finish_reason": finish_reason
    }


print("Reusable BM25 answer-generation function created successfully.")
```

## Configuration B: Evidence Selection Score Diagnosis

The first improved evidence-selection attempt did not
prioritise several passages containing explicit machine
learning methods.

This step examines all 20 BM25 candidates, including their
BM25 scores, question-term overlap, combined selection
scores, and passage lengths.

The purpose is to identify why useful passages were
excluded before refining the selection strategy.

```python
# Recalculate the diagnostic scores using the existing function's
# candidate pool, without changing the original BM25 index.

q001_score_diagnostic = bm25_search_deduplicated(
    q001_question,
    top_k=20
).copy()

q001_score_diagnostic = q001_score_diagnostic.reset_index(drop=True)
q001_score_diagnostic["bm25_rank"] = (
    q001_score_diagnostic.index + 1
)

# Reproduce the question-term overlap calculation
import re

stopwords = {
    "what", "which", "who", "where", "when", "why", "how",
    "does", "did", "the", "a", "an", "and", "or", "of",
    "in", "on", "for", "to", "from", "with", "by",
    "their", "they", "this", "that", "these", "those",
    "used", "use", "using", "are", "is", "was", "were"
}

question_terms = {
    term.lower()
    for term in re.findall(r"\b[a-zA-Z0-9]+\b", q001_question)
    if len(term) > 2 and term.lower() not in stopwords
}

def calculate_q001_overlap(text):
    passage_terms = {
        term.lower()
        for term in re.findall(r"\b[a-zA-Z0-9]+\b", str(text))
    }

    if not question_terms:
        return 0.0

    return (
        len(question_terms.intersection(passage_terms))
        / len(question_terms)
    )

q001_score_diagnostic["question_overlap"] = (
    q001_score_diagnostic["text"].apply(calculate_q001_overlap)
)

# Reproduce the normalised BM25 and combined scores
score_min = q001_score_diagnostic["retrieval_score"].min()
score_max = q001_score_diagnostic["retrieval_score"].max()

if score_max > score_min:
    q001_score_diagnostic["normalised_bm25"] = (
        (q001_score_diagnostic["retrieval_score"] - score_min)
        / (score_max - score_min)
    )
else:
    q001_score_diagnostic["normalised_bm25"] = 1.0

q001_score_diagnostic["selection_score"] = (
    0.65 * q001_score_diagnostic["normalised_bm25"]
    + 0.35 * q001_score_diagnostic["question_overlap"]
)

q001_score_diagnostic["passage_length"] = (
    q001_score_diagnostic["text"].astype(str).str.len()
)

# Display all candidates ordered by selection score
diagnostic_columns = [
    "bm25_rank",
    "article_id",
    "page_number",
    "retrieval_score",
    "question_overlap",
    "normalised_bm25",
    "selection_score",
    "passage_length"
]

display(
    q001_score_diagnostic
    .sort_values("selection_score", ascending=False)[diagnostic_columns]
    .reset_index(drop=True)
)

print("\nQuestion terms:", sorted(question_terms))
print("Total candidates:", len(q001_score_diagnostic))
print("Total candidate characters:",
      q001_score_diagnostic["passage_length"].sum())
```

```python
import re
import pandas as pd

# Recognised methods and model names relevant to the project domain.
# These terms are used as evidence indicators, not as proof of a
# method's exact role in a research study.
METHOD_TERMS = [
    "svm",
    "support vector machine",
    "knn",
    "k-nn",
    "k nearest neighbor",
    "random forest",
    "decision tree",
    "decision trees",
    "cnn",
    "convolutional neural network",
    "cnn-lstm",
    "lstm",
    "dnn",
    "deep neural network",
    "ann",
    "artificial neural network",
    "brann",
    "k-means",
    "mobilenet",
    "mobilenetv2",
    "resnet",
    "resnet50",
    "vgg",
    "inception",
    "densenet",
    "transformer",
    "vision transformer",
    "capsnet",
    "naive bayes",
    "logistic regression",
    "xgboost",
    "gradient boosting",
    "random forest classifier"
]


def count_method_evidence(text):
    """Count distinct recognised method terms in a passage."""

    text = str(text).lower()
    found_methods = set()

    for method in METHOD_TERMS:
        pattern = r"(?<!\w)" + re.escape(method) + r"(?!\w)"

        if re.search(pattern, text):
            found_methods.add(method)

    return len(found_methods), sorted(found_methods)


def retrieve_select_bm25_evidence(
    question,
    candidate_k=20,
    max_passages=5,
    max_characters=6500
):
    """
    Retrieve BM25 candidates and select evidence using
    method coverage for method-related questions.

    This is a lexical/rule-based heuristic, not a
    transformer-based reranker.
    """

    candidates = bm25_search_deduplicated(
        question,
        top_k=candidate_k
    ).copy()

    if candidates.empty:
        return pd.DataFrame()

    candidates = candidates.reset_index(drop=True)
    candidates["bm25_rank"] = candidates.index + 1

    question_lower = question.lower()

    method_question = any(
        term in question_lower
        for term in [
            "method", "methods", "technique", "techniques",
            "algorithm", "algorithms", "model", "models",
            "approach", "approaches"
        ]
    )

    candidates["method_count"] = 0
    candidates["methods_found"] = None

    if method_question:
        method_results = candidates["text"].apply(
            count_method_evidence
        )

        candidates["method_count"] = method_results.apply(
            lambda result: result[0]
        )

        candidates["methods_found"] = method_results.apply(
            lambda result: ", ".join(result[1])
        )

        # Prioritise passages with more method evidence.
        candidates = candidates.sort_values(
            by=["method_count", "retrieval_score"],
            ascending=[False, False]
        )

    else:
        # Retain ordinary BM25 ranking for other question types.
        candidates = candidates.sort_values(
            by="retrieval_score",
            ascending=False
        )

    selected = []
    used_characters = 0

    for _, row in candidates.iterrows():

        passage = str(row["text"]).strip()

        if not passage:
            continue

        # Keep complete passages; do not truncate evidence.
        if used_characters + len(passage) > max_characters:
            continue

        selected.append(row)
        used_characters += len(passage)

        if len(selected) >= max_passages:
            break

    return pd.DataFrame(selected).reset_index(drop=True)


print("Method-aware BM25 evidence selector created successfully.")
```

```python
# Apply the method-aware selector to Q001
q001_method_evidence = retrieve_select_bm25_evidence(
    question=q001_question,
    candidate_k=20,
    max_passages=5,
    max_characters=6500
)

print("=" * 85)
print("Q001 – METHOD-AWARE SELECTED EVIDENCE")
print("=" * 85)

for rank, (_, row) in enumerate(
    q001_method_evidence.iterrows(),
    start=1
):
    print(f"\nSelected passage {rank}")
    print("Original BM25 rank:", row["bm25_rank"])
    print("Article ID:", row["article_id"])
    print("Title:", row["title"])
    print("Page:", row["page_number"])
    print("Detected method count:", row["method_count"])
    print("Detected methods:", row["methods_found"])
    print("Passage:", str(row["text"])[:800])

print("\nTotal selected passages:", len(q001_method_evidence))
```

```python
import re
import time

# Extract concise sentences containing recognised methods
focused_evidence = []

for i, (_, row) in enumerate(q001_method_evidence.iterrows(), start=1):

    source_label = f"S{i}"
    passage = str(row["text"])

    sentences = re.split(r"(?<=[.!?])\s+", passage)

    relevant_sentences = []

    for sentence in sentences:
        sentence_lower = sentence.lower()

        if any(
            re.search(
                r"(?<!\w)" + re.escape(method) + r"(?!\w)",
                sentence_lower
            )
            for method in METHOD_TERMS
        ):
            relevant_sentences.append(sentence.strip())

    # Keep at most two relevant sentences per paper
    selected_sentences = relevant_sentences[:2]

    if selected_sentences:
        focused_evidence.append(
            f"[{source_label}] {row['article_id']}, "
            f"page {row['page_number']}: "
            + " ".join(selected_sentences)
        )

# Build a short prompt without the repeated abstention instruction
compact_context = "\n".join(focused_evidence)

q001_short_prompt = f"""Question: {q001_question}

Use these research-paper evidence excerpts:

{compact_context}

List the machine-learning and deep-learning methods explicitly
mentioned in the excerpts. Keep the answer to two sentences.
Cite each sentence using the source labels [S1], [S2], or [S3].
Do not repeat the question. Do not say evidence is insufficient.

Answer:"""

# Check prompt size before generation
prompt_tokens = len(
    llm.tokenize(q001_short_prompt.encode("utf-8"))
)

print("Prompt tokens:", prompt_tokens)
print("\nFocused evidence:")
print(compact_context)

if prompt_tokens > 1700:
    print("\nPrompt is too long. Generation skipped.")
else:
    start_time = time.time()

    q001_short_response = llm(
        q001_short_prompt,
        max_tokens=100,
        temperature=0.0,
        repeat_penalty=1.15,
        echo=False
    )

    q001_short_response_time = time.time() - start_time

    q001_short_answer = (
        q001_short_response["choices"][0]["text"].strip()
    )

    print("\nGenerated Q001 answer:")
    print(q001_short_answer)

    print(
        "\nResponse time:",
        round(q001_short_response_time, 3),
        "seconds"
    )

    print(
        "Finish reason:",
        q001_short_response["choices"][0]["finish_reason"]
    )
```

```python
q001_final_prompt = f"""Question: {q001_question}

Evidence:
{compact_context}

Answer in exactly two short sentences.
Identify the classification methods and deep-learning
methods. Mention feature extraction separately if relevant.
Do not include explainability methods.
Cite the evidence using [S1], [S2], or [S3].
Use only the evidence above.

Answer:"""

start_time = time.time()

q001_final_response = llm(
    q001_final_prompt,
    max_tokens=160,
    temperature=0.0,
    repeat_penalty=1.15
)

q001_final_time = time.time() - start_time

q001_final_answer = (
    q001_final_response["choices"][0]["text"].strip()
)

print("Q001:", q001_question)
print("\nFinal generated answer:")
print(q001_final_answer)

print("\nResponse time:", round(q001_final_time, 3), "seconds")
print("Finish reason:", q001_final_response["choices"][0]["finish_reason"])
```

```python
def select_configuration_b_evidence(question_row):

    question_id = question_row["question_id"]
    category = question_row["category"]
    question = question_row["question"]

    if (
        category == "Factual"
        and any(
            word in question.lower()
            for word in [
                "method", "methods", "algorithm", "algorithms",
                "technique", "techniques"
            ]
        )
    ):
        evidence = retrieve_select_bm25_evidence(
            question=question,
            candidate_k=20,
            max_passages=5,
            max_characters=6500
        )

        selection_type = "Method-aware BM25"

    else:
        evidence = bm25_search_deduplicated(
            question,
            top_k=5
        ).copy()

        evidence["bm25_rank"] = range(1, len(evidence) + 1)
        evidence["method_count"] = 0
        evidence["methods_found"] = ""

        selection_type = "Original BM25 top-5"

    return evidence, selection_type


print("Category-aware Configuration B selector created successfully.")
```

## Configuration B: Validate Retrieval for 80 Questions

This step applies the category-aware BM25 evidence
selector to all 80 evaluation questions.

It records the selected passages, retrieval strategy,
article identifiers, page numbers, and original BM25 ranks.

The results will be saved for use in answer generation
and subsequent retrieval evaluation.

```python
# Store evidence and retrieval information for all questions
configuration_b_retrieval = {}
retrieval_summary = []

for _, question_row in evaluation_df.iterrows():

    question_id = question_row["question_id"]

    evidence, selection_type = select_configuration_b_evidence(
        question_row
    )

    configuration_b_retrieval[question_id] = evidence

    retrieval_summary.append({
        "question_id": question_id,
        "category": question_row["category"],
        "selection_type": selection_type,
        "number_of_passages": len(evidence),
        "retrieved_article_ids": (
            ", ".join(evidence["article_id"].astype(str))
            if not evidence.empty else ""
        ),
        "retrieved_pages": (
            ", ".join(evidence["page_number"].astype(str))
            if not evidence.empty else ""
        )
    })

retrieval_summary_df = pd.DataFrame(retrieval_summary)

print("Total questions processed:", len(retrieval_summary_df))
print("\nRetrieval strategy distribution:")
print(retrieval_summary_df["selection_type"].value_counts())

print("\nPassage count distribution:")
print(retrieval_summary_df["number_of_passages"].value_counts().sort_index())

print("\nQuestions without retrieved evidence:")
print(
    retrieval_summary_df.loc[
        retrieval_summary_df["number_of_passages"] == 0,
        "question_id"
    ].tolist()
)

display(retrieval_summary_df.head(10))
```

## Configuration B: Generate All 80 Answers

This step generates answers for the complete set of
80 evaluation questions using BM25-retrieved evidence
and the Qwen2.5-0.5B-Instruct language model.

Each question receives its selected passages with
source identifiers and page metadata.

The model is instructed to answer only from retrieved
evidence and abstain when the evidence is insufficient.

Generated answers, citations, response times, and
completion reasons are recorded without manual editing.

```python
import time
import json
import pandas as pd
from pathlib import Path

configuration_b_records = []
configuration_b_source_maps = {}

for index, (_, question_row) in enumerate(
    evaluation_df.iterrows(), start=1
):

    question_id = question_row["question_id"]
    question = question_row["question"]

    evidence = configuration_b_retrieval[question_id]

    # Prepare source-labelled evidence
    source_map = {}
    context_parts = []

    # Start with the highest-ranked selected passages.
    # Reduce the number of passages if the prompt is too large.
    for _, row in evidence.iterrows():

        label = f"S{len(source_map) + 1}"

        source_map[label] = {
            "article_id": row["article_id"],
            "title": row["title"],
            "page_number": row["page_number"],
            "chunk_id": row["chunk_id"]
        }

        context_parts.append(
            f"[{label}] Article: {row['title']}\n"
            f"Article ID: {row['article_id']}\n"
            f"Page: {row['page_number']}\n"
            f"Evidence: {row['text']}"
        )

    # Build a prompt that fits the available model context.
    while context_parts:

        prompt = f"""You are an academic research assistant.

Question:
{question}

Retrieved research evidence:
{chr(10).join(context_parts)}

Answer the question concisely using only the evidence.
Cite supporting statements with [S1], [S2], etc.
Do not invent facts or citations.
If the evidence does not answer the question, state:
Insufficient evidence in the retrieved papers.

Answer:"""

        prompt_tokens = len(
            llm.tokenize(prompt.encode("utf-8"))
        )

        if prompt_tokens <= 1650:
            break

        # Remove the lowest-priority selected passage.
        context_parts.pop()
        source_map.pop(f"S{len(source_map)}", None)

    if not context_parts:
        generated_answer = (
            "Insufficient evidence in the retrieved papers."
        )
        response_time = 0.0
        finish_reason = "skipped_context_limit"

    else:
        start_time = time.time()

        response = llm(
            prompt,
            max_tokens=120,
            temperature=0.0,
            repeat_penalty=1.15
        )

        response_time = time.time() - start_time

        generated_answer = (
            response["choices"][0]["text"].strip()
        )

        finish_reason = response["choices"][0]["finish_reason"]

    configuration_b_source_maps[question_id] = source_map

    configuration_b_records.append({
        "question_id": question_id,
        "category": question_row["category"],
        "question": question,
        "gold_answer": question_row["gold_answer"],
        "answerable": question_row["answerable"],
        "supporting_sources": question_row["supporting_sources"],
        "generated_answer": generated_answer,
        "response_time": response_time,
        "finish_reason": finish_reason,
        "prompt_tokens": prompt_tokens if context_parts else 0,
        "retrieved_source_count": len(source_map),
        "source_map": json.dumps(source_map, ensure_ascii=False)
    })

    print(
        f"{index}/80 | {question_id} | "
        f"{response_time:.2f}s | {finish_reason}"
    )

# Create the final results DataFrame
configuration_b_results_df = pd.DataFrame(
    configuration_b_records
)

# Save results
EVALUATION_DIR = Path(
    "/Users/macbookair/DAT401_Research_Assistant/evaluation"
)

results_path = (
    EVALUATION_DIR / "configuration_B_bm25_llm_results.csv"
)

configuration_b_results_df.to_csv(
    results_path,
    index=False,
    encoding="utf-8-sig"
)

# Save source mappings separately
source_map_path = (
    EVALUATION_DIR / "configuration_B_source_maps.json"
)

with open(source_map_path, "w", encoding="utf-8") as file:
    json.dump(
        configuration_b_source_maps,
        file,
        indent=4,
        ensure_ascii=False
    )

print("\n" + "=" * 70)
print("CONFIGURATION B GENERATION SUMMARY")
print("=" * 70)

print("Total questions:", len(configuration_b_results_df))
print("Unique question IDs:",
      configuration_b_results_df["question_id"].nunique())
print("Missing answers:",
      configuration_b_results_df["generated_answer"].isna().sum())

print("\nCompletion reasons:")
print(configuration_b_results_df["finish_reason"].value_counts())

print("\nAverage response time:",
      round(configuration_b_results_df["response_time"].mean(), 3),
      "seconds")

print("\nResults saved:", results_path)
print("Source mappings saved:", source_map_path)
```

- Missing or empty answers.

- Citation labels in generated answers.

## Configuration B: Validate the Revised Generated Answers

## Configuration B: Automated First-Pass Answer Diagnostics

This step examines all 80 generated answers to identify possible evaluation issues.

It checks:
- Whether answers contain an abstention statement.
- Whether citation markers appear in the generated text.
- Whether source information is available.
- Whether the response reached the generation token limit.
- Whether answers are unusually short or long.

These are diagnostic indicators, not final correctness or faithfulness scores. Citation markers must still be checked against their actual sources.

```python
import re
import pandas as pd

# Work on a copy so the original generated results remain unchanged
b_diagnostics_df = configuration_b_results_400_df.copy()

# Identify likely abstention statements
abstention_pattern = re.compile(
    r"insufficient evidence|cannot determine|cannot answer|"
    r"not enough information|unable to answer|not provided in the",
    flags=re.IGNORECASE
)

# Identify common citation formats, such as [S1] or [S2]
citation_pattern = re.compile(r"\[S\d+\]", flags=re.IGNORECASE)

# Check whether a value contains usable source information
def has_source_information(value):
    if isinstance(value, (list, dict)):
        return len(value) > 0

    if pd.isna(value):
        return False

    value = str(value).strip()
    return value not in ["", "[]", "{}", "nan", "None"]

# Create diagnostic indicators for every answer
b_diagnostics_df["contains_abstention"] = (
    b_diagnostics_df["generated_answer"]
    .fillna("")
    .apply(lambda x: bool(abstention_pattern.search(str(x))))
)

b_diagnostics_df["contains_citation_marker"] = (
    b_diagnostics_df["generated_answer"]
    .fillna("")
    .apply(lambda x: bool(citation_pattern.search(str(x))))
)

b_diagnostics_df["has_source_information"] = (
    b_diagnostics_df["source_map"]
    .apply(has_source_information)
)

b_diagnostics_df["generation_truncated"] = (
    b_diagnostics_df["finish_reason"]
    .astype(str)
    .str.lower()
    .eq("length")
)

# Summarise all 80 answers
print("CONFIGURATION B: AUTOMATED DIAGNOSTIC SUMMARY")
print("-" * 50)
print("Total answers:", len(b_diagnostics_df))
print("Answers containing abstention language:",
      b_diagnostics_df["contains_abstention"].sum())
print("Answers containing citation markers:",
      b_diagnostics_df["contains_citation_marker"].sum())
print("Answers with source information:",
      b_diagnostics_df["has_source_information"].sum())
print("Answers reaching token limit:",
      b_diagnostics_df["generation_truncated"].sum())

print("\nDiagnostic counts by question category:")
display(
    b_diagnostics_df.groupby("category")[
        [
            "contains_abstention",
            "contains_citation_marker",
            "generation_truncated"
        ]
    ].sum()
)

print("\nSample of flagged answers:")
display(
    b_diagnostics_df.loc[
        b_diagnostics_df["generation_truncated"]
        | ~b_diagnostics_df["contains_citation_marker"],
        [
            "question_id",
            "category",
            "answerable",
            "finish_reason",
            "contains_abstention",
            "contains_citation_marker",
            "generated_answer"
        ]
    ].head(10)
)
```

## Configuration B: Calculate BM25 Retrieval Metrics

Retrieve the top ten passages for each evaluation question using the existing deduplicated BM25 index.

Match retrieved papers against the expected relevant paper titles and calculate:

- Recall@5: proportion of expected relevant papers retrieved within the top five.
- MRR@10: reciprocal rank of the first relevant paper within the top ten.

Only answerable questions with successfully matched expected sources are included in the metric calculations. Unmatched expected titles will be reported separately.

```python
import re
import pandas as pd
import unicodedata
from pathlib import Path

# Standardise paper titles for matching
def normalise_title(title):
    title = unicodedata.normalize("NFKD", str(title))
    title = title.lower()
    title = re.sub(r"[^a-z0-9]+", " ", title)
    return " ".join(title.split())


# Map normalised corpus titles to article IDs
title_to_article_ids = {}

for _, row in bm25_deduplicated_df[
    ["article_id", "title"]
].drop_duplicates().iterrows():

    key = normalise_title(row["title"])
    title_to_article_ids.setdefault(key, set()).add(row["article_id"])


# Match expected relevant titles to corpus article IDs
expected_article_ids = {}
unmatched_titles = []

for question_id, expected_titles in corrected_expected_sources.items():

    matched_ids = set()

    for title in expected_titles:
        key = normalise_title(title)

        if key in title_to_article_ids:
            matched_ids.update(title_to_article_ids[key])
        else:
            unmatched_titles.append({
                "question_id": question_id,
                "expected_title": title
            })

    expected_article_ids[question_id] = matched_ids


# Retrieve top 10 passages for all 80 questions
retrieval_records = []
metric_records = []

for _, question_row in evaluation_df.iterrows():

    question_id = question_row["question_id"]
    question = question_row["question"]
    answerable = bool(question_row["answerable"])

    results = bm25_search_deduplicated(
        question,
        top_k=10
    )

    relevant_ids = expected_article_ids.get(question_id, set())

    retrieved_ids = (
        results["article_id"].tolist()
        if not results.empty else []
    )

    # Save ranked retrieval results
    for rank, (_, result) in enumerate(results.iterrows(), start=1):
        retrieval_records.append({
            "question_id": question_id,
            "category": question_row["category"],
            "rank": rank,
            "article_id": result["article_id"],
            "title": result["title"],
            "page_number": result["page_number"],
            "retrieval_score": result["retrieval_score"]
        })

    # Calculate metrics only for answerable questions
    # with matched expected sources
    if answerable and relevant_ids:

        relevant_retrieved_top5 = (
            set(retrieved_ids[:5]) & relevant_ids
        )

        recall_at_5 = (
            len(relevant_retrieved_top5) / len(relevant_ids)
        )

        first_relevant_rank = next(
            (
                rank
                for rank, article_id in enumerate(
                    retrieved_ids, start=1
                )
                if article_id in relevant_ids
            ),
            None
        )

        reciprocal_rank_at_10 = (
            1 / first_relevant_rank
            if first_relevant_rank is not None
            else 0
        )

        metric_records.append({
            "question_id": question_id,
            "category": question_row["category"],
            "relevant_paper_count": len(relevant_ids),
            "relevant_papers_retrieved_top5": len(
                relevant_retrieved_top5
            ),
            "recall_at_5": recall_at_5,
            "first_relevant_rank": first_relevant_rank,
            "reciprocal_rank_at_10": reciprocal_rank_at_10
        })


# Create evaluation DataFrames
bm25_top10_results_df = pd.DataFrame(retrieval_records)
bm25_final_retrieval_evaluation_df = pd.DataFrame(metric_records)

# Calculate overall metrics
recall_at_5_mean = (
    bm25_final_retrieval_evaluation_df["recall_at_5"].mean()
)

mrr_at_10_mean = (
    bm25_final_retrieval_evaluation_df[
        "reciprocal_rank_at_10"
    ].mean()
)

# Display evaluation coverage and metrics
print("CONFIGURATION B: BM25 RETRIEVAL EVALUATION")
print("-" * 50)

print("Total evaluation questions:", len(evaluation_df))
print("Top-10 retrieval records:", len(bm25_top10_results_df))
print("Answerable questions evaluated:",
      len(bm25_final_retrieval_evaluation_df))
print("Unmatched expected titles:", len(unmatched_titles))

print(f"\nRecall@5: {recall_at_5_mean:.4f} "
      f"({recall_at_5_mean * 100:.2f}%)")

print(f"MRR@10: {mrr_at_10_mean:.4f}")

print("\nMetrics by category:")
display(
    bm25_final_retrieval_evaluation_df
    .groupby("category")[
        ["recall_at_5", "reciprocal_rank_at_10"]
    ]
    .mean()
)

if unmatched_titles:
    print("\nExpected titles not matched to corpus:")
    display(pd.DataFrame(unmatched_titles))

# Save the results
evaluation_folder = Path(
    "/Users/macbookair/DAT401_Research_Assistant/evaluation"
)

bm25_top10_results_df.to_csv(
    evaluation_folder / "configuration_B_bm25_top10_results.csv",
    index=False
)

bm25_final_retrieval_evaluation_df.to_csv(
    evaluation_folder / "configuration_B_retrieval_metrics.csv",
    index=False
)

print("\nRetrieval evaluation files saved successfully.")
```

## Configuration B: Preliminary Answer Evaluation

Prepare preliminary answer-quality scores using automated checks for answer content, relevance, citation references, and abstention behaviour.

These scores are diagnostic estimates. Citation support and factual correctness require human verification before final reporting.

```python
import re
import pandas as pd

# Work on a copy of the existing evaluation sheet
configuration_b_evaluation_df = configuration_b_evaluation_df.copy()

def evaluate_preliminary(row):

    answer = str(row["generated_answer"])
    answer_lower = answer.lower()

    question = str(row["question"])
    answerable = bool(row["answerable"])

    # Identify abstention language
    abstention_phrases = [
        "insufficient evidence",
        "cannot answer",
        "not enough information",
        "unable to determine"
    ]

    contains_abstention = any(
        phrase in answer_lower
        for phrase in abstention_phrases
    )

    # Identify whether the answer contains a substantive response
    # Remove repeated evidence headings and citation labels
    cleaned_answer = re.sub(
        r"\[S\d+\]",
        "",
        answer,
        flags=re.IGNORECASE
    )

    cleaned_answer = re.sub(
        r"retrieved research evidence:.*",
        "",
        cleaned_answer,
        flags=re.IGNORECASE | re.DOTALL
    )

    substantive_text = cleaned_answer.strip()

    # Check source labels against the question's source map
    source_map = row["source_map"]

    if isinstance(source_map, dict):
        valid_labels = set(source_map.keys())
    else:
        valid_labels = set()

    cited_labels = set(
        re.findall(r"\[S\d+\]", answer, flags=re.IGNORECASE)
    )

    resolved_labels = {
        label.upper()
        for label in cited_labels
        if label.upper() in {
            key.upper() for key in valid_labels
        }
    }

    # Preliminary citation score:
    # 0 = no citation or unresolved citation
    # 1 = citation resolves to a source
    # 2 = source support verified manually (not automated here)
    citation_score = (
        1 if resolved_labels else 0
    )

    # Preliminary relevance
    if not substantive_text:
        relevance_score = 0
    elif contains_abstention:
        relevance_score = 1
    else:
        relevance_score = 2

    # Correctness and faithfulness are deliberately left blank.
    # They cannot be reliably established by keyword checks.
    correctness_score = None
    faithfulness_score = None

    # Abstention correctness is assessed only for unanswerable questions.
    if not answerable:
        abstention_correct = (
            2 if contains_abstention else 0
        )
    else:
        abstention_correct = None

    return pd.Series({
        "correctness_score": correctness_score,
        "relevance_score": relevance_score,
        "faithfulness_score": faithfulness_score,
        "citation_validity": citation_score,
        "abstention_correct": abstention_correct,
        "review_notes": (
            "Provisional automated diagnostic; "
            "factual correctness and citation support need manual verification."
        )
    })


# Apply preliminary diagnostics
preliminary_scores = configuration_b_evaluation_df.apply(
    evaluate_preliminary,
    axis=1
)

for column in preliminary_scores.columns:
    configuration_b_evaluation_df[column] = preliminary_scores[column]


# Display diagnostic summary
print("PRELIMINARY CONFIGURATION B EVALUATION")
print("-" * 50)

print("Total questions:", len(configuration_b_evaluation_df))

print("\nRelevance score distribution:")
display(
    configuration_b_evaluation_df[
        "relevance_score"
    ].value_counts(dropna=False).sort_index()
)

print("\nCitation resolution score distribution:")
display(
    configuration_b_evaluation_df[
        "citation_validity"
    ].value_counts(dropna=False).sort_index()
)

print("\nUnanswerable-question abstention results:")
display(
    configuration_b_evaluation_df[
        configuration_b_evaluation_df["answerable"] == False
    ]["abstention_correct"].value_counts(dropna=False)
)

# Save a separate provisional evaluation file
output_path = (
    "/Users/macbookair/DAT401_Research_Assistant/evaluation/"
    "configuration_B_preliminary_evaluation.csv"
)

configuration_b_evaluation_df.to_csv(
    output_path,
    index=False
)

print("\nSaved:", output_path)
print("\nCorrectness and faithfulness remain unscored pending verification.")
```

```python
import json
import re
import pandas as pd

def check_citation_resolution(row):

    answer = str(row["generated_answer"])
    raw_source_map = row["source_map"]

    # Convert JSON string into a Python dictionary
    try:
        source_map = (
            json.loads(raw_source_map)
            if isinstance(raw_source_map, str)
            else raw_source_map
        )
    except (json.JSONDecodeError, TypeError):
        source_map = {}

    # Extract citation labels from the generated answer
    cited_labels = {
        label.upper()
        for label in re.findall(
            r"\[S\d+\]", answer, flags=re.IGNORECASE
        )
    }

    # Compare citation labels with available source-map keys
    available_labels = {
        str(label).upper()
        for label in source_map.keys()
    } if isinstance(source_map, dict) else set()

    resolved_labels = cited_labels.intersection(available_labels)

    if not cited_labels:
        return pd.Series({
            "citation_validity": 0,
            "citation_check": "No citation labels"
        })

    if resolved_labels == cited_labels:
        return pd.Series({
            "citation_validity": 1,
            "citation_check": "All citation labels resolve to sources"
        })

    return pd.Series({
        "citation_validity": 0,
        "citation_check": "Some citation labels are unresolved"
    })


# Apply corrected citation checks
citation_checks = configuration_b_evaluation_df.apply(
    check_citation_resolution,
    axis=1
)

configuration_b_evaluation_df["citation_validity"] = (
    citation_checks["citation_validity"]
)

configuration_b_evaluation_df["citation_check"] = (
    citation_checks["citation_check"]
)

print("CORRECTED CITATION RESOLUTION")
print("-" * 45)

display(
    configuration_b_evaluation_df[
        ["citation_validity", "citation_check"]
    ].value_counts().to_frame("Number of questions")
)

print("\nExamples:")
display(
    configuration_b_evaluation_df[
        ["question_id", "citation_validity", "citation_check"]
    ].head(10)
)
```

```python
import json
import re
import pandas as pd

def check_citation_resolution(row):

    answer = str(row["generated_answer"])
    raw_map = row["source_map"]

    try:
        source_map = (
            json.loads(raw_map)
            if isinstance(raw_map, str)
            else raw_map
        )
    except (json.JSONDecodeError, TypeError):
        source_map = {}

    # Extract S1, S2, S3 without square brackets
    cited_labels = {
        label.upper()
        for label in re.findall(
            r"\[(S\d+)\]",
            answer,
            flags=re.IGNORECASE
        )
    }

    available_labels = {
        str(label).upper()
        for label in source_map.keys()
    } if isinstance(source_map, dict) else set()

    resolved = cited_labels.intersection(available_labels)

    if not cited_labels:
        status = "No citation labels"
        score = 0
    elif resolved == cited_labels:
        status = "All labels resolve"
        score = 1
    else:
        status = "Some labels unresolved"
        score = 0

    return pd.Series({
        "citation_validity": score,
        "citation_check": status,
        "citation_count": len(cited_labels),
        "resolved_citation_count": len(resolved)
    })


citation_checks = configuration_b_results_400_df.apply(
    check_citation_resolution,
    axis=1
)

for column in citation_checks.columns:
    configuration_b_evaluation_df[column] = citation_checks[column]

print("FINAL CITATION RESOLUTION CHECK")
print("-" * 45)

print(
    configuration_b_evaluation_df["citation_check"]
    .value_counts()
)

print("\nExamples:")
display(
    configuration_b_evaluation_df[
        [
            "question_id",
            "citation_count",
            "resolved_citation_count",
            "citation_validity",
            "citation_check"
        ]
    ].head(10)
)
```

```python
import re
import pandas as pd

def inspect_answer(row):

    answer = str(row["generated_answer"])

    # Remove the appended retrieved-evidence section
    answer_part = re.split(
        r"Retrieved research evidence:",
        answer,
        flags=re.IGNORECASE
    )[0].strip()

    # Count abstention statements
    abstention_count = len(re.findall(
        r"insufficient evidence|cannot answer|not enough information",
        answer_part,
        flags=re.IGNORECASE
    ))

    # Remove common abstention wording for inspection
    remaining_text = re.sub(
        r"insufficient evidence in the retrieved papers\.?",
        "",
        answer_part,
        flags=re.IGNORECASE
    ).strip()

    return pd.Series({
        "answer_text": answer_part,
        "answer_characters": len(answer_part),
        "abstention_count": abstention_count,
        "has_non_abstention_text": len(remaining_text) > 30
    })


answer_inspection = configuration_b_results_400_df.apply(
    inspect_answer,
    axis=1
)

configuration_b_inspection_df = pd.concat(
    [
        configuration_b_results_400_df[
            ["question_id", "category", "question", "gold_answer", "answerable"]
        ],
        answer_inspection
    ],
    axis=1
)

print("ANSWER INSPECTION SUMMARY")
print("-" * 45)

print("Total answers:", len(configuration_b_inspection_df))
print(
    "Answers with additional text:",
    configuration_b_inspection_df[
        "has_non_abstention_text"
    ].sum()
)
print(
    "Answers with no additional text:",
    (~configuration_b_inspection_df[
        "has_non_abstention_text"
    ]).sum()
)

print("\nSample answers for assessment:")
display(
    configuration_b_inspection_df[
        [
            "question_id",
            "category",
            "answerable",
            "abstention_count",
            "answer_text"
        ]
    ].head(12)
)

# Save the inspection results
configuration_b_inspection_df.to_csv(
    "/Users/macbookair/DAT401_Research_Assistant/evaluation/"
    "configuration_B_answer_inspection.csv",
    index=False
)
```

### Updating Configuration B Evaluation Scores

The dense retrieval dataset contains research passages extracted

Each passage includes article metadata, page information,
section information and passage text.

This dataset will be used as the document collection for

```python
import pandas as pd
import numpy as np
from pathlib import Path

PROJECT_DIR = Path(
    "/Users/macbookair/DAT401_Research_Assistant"
)

EVALUATION_DIR = PROJECT_DIR / "evaluation"

DENSE_DATA_PATH = (
    EVALUATION_DIR / "dense_retrieval_dataset.csv"
)

dense_retrieval_df = pd.read_csv(DENSE_DATA_PATH)

print("DENSE RETRIEVAL DATASET")
print("=" * 50)

print("Dataset shape:", dense_retrieval_df.shape)

print("\nColumns:")
print(dense_retrieval_df.columns.tolist())

print("\nMissing values:")
print(dense_retrieval_df.isnull().sum())

print("\nSample passages:")
display(dense_retrieval_df.head(3))
```

## 2. Load and Validate Document Embeddings

Document embeddings represent each research passage as a numerical
vector. These vectors allow semantic similarity comparison between
a user's question and the stored research passages.

The embedding matrix should contain one vector for every passage
in the dense retrieval dataset.

```python
DENSE_EMBEDDINGS_PATH = (
    EVALUATION_DIR / "dense_document_embeddings.npy"
)

dense_document_embeddings = np.load(DENSE_EMBEDDINGS_PATH)

print("DOCUMENT EMBEDDINGS")
print("=" * 50)

print("Embedding shape:", dense_document_embeddings.shape)
print("Number of passages:", len(dense_retrieval_df))
print("Number of embedding vectors:", len(dense_document_embeddings))

print("\nData type:", dense_document_embeddings.dtype)

print("\nMissing or invalid values:",
      not np.isfinite(dense_document_embeddings).all())

print("\nAll embeddings finite:",
      np.isfinite(dense_document_embeddings).all())

print("\nFirst embedding (first 10 values):")
print(dense_document_embeddings[0][:10])
```

## 3. Load the Sentence Transformer Model

The Sentence Transformer model converts research questions into
384-dimensional numerical embeddings.

The same model used to generate the document embeddings must be
used for query embeddings to ensure compatibility.

```python
from sentence_transformers import SentenceTransformer

EMBEDDING_MODEL_NAME = "sentence-transformers/all-MiniLM-L6-v2"

embedding_model_eval = SentenceTransformer(EMBEDDING_MODEL_NAME)

print("Embedding model loaded successfully!")

print("\nModel name:", EMBEDDING_MODEL_NAME)

print(
    "Embedding dimension:",
    embedding_model_eval.get_sentence_embedding_dimension()
)
```

## 4. Test Query Embedding Generation

Convert a sample research question into a numerical embedding.

The query embedding must have 384 dimensions to match the
document embeddings.

```python
sample_query = "What machine learning methods are used for brain tumour classification?"

query_embedding = embedding_model_eval.encode(
    [sample_query],
    convert_to_numpy=True,
    normalize_embeddings=True
)

print("QUERY EMBEDDING TEST")
print("=" * 50)

print("Query:", sample_query)
print("Query embedding shape:", query_embedding.shape)
print("Document embedding shape:", dense_document_embeddings.shape)

print(
    "Dimensions match:",
    query_embedding.shape[1] == dense_document_embeddings.shape[1]
)

print("\nFirst 10 values:")
print(query_embedding[0][:10])
```

## 5. Implement Dense Retrieval

Dense retrieval uses semantic similarity to identify research
passages relevant to a given question.

Cosine similarity measures the similarity between the query
embedding and document embeddings. Passages with higher scores
are ranked as more relevant.

```python
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np

def dense_search_eval(query, top_k=5):

    # Convert the question into an embedding
    query_embedding = embedding_model_eval.encode(
        [query],
        convert_to_numpy=True,
        normalize_embeddings=True
    )

    # Calculate cosine similarity with all document embeddings
    scores = cosine_similarity(
        query_embedding,
        dense_document_embeddings
    )[0]

    # Select the passages with the highest similarity scores
    top_indices = np.argsort(scores)[::-1][:top_k]

    # Retrieve the corresponding passages
    results = dense_retrieval_df.iloc[top_indices].copy()

    # Add similarity scores and ranking
    results["retrieval_score"] = scores[top_indices]
    results["rank"] = range(1, len(results) + 1)

    return results


print("Dense retrieval function defined successfully!")
```

## 6. Test Dense Retrieval

Test the dense retrieval function using a sample research question.

The function should return the five passages with the highest
cosine similarity scores, ranked from highest to lowest.

```python
sample_query = "What machine learning methods are used for brain tumour classification?"

test_results = dense_search_eval(sample_query, top_k=5)

print("DENSE RETRIEVAL TEST")
print("=" * 70)

for _, row in test_results.iterrows():
    print(f"\nRank: {row['rank']}")
    print(f"Article ID: {row['article_id']}")
    print(f"Title: {row['title']}")
    print(f"Page: {row['page_number']}")
    print(f"Similarity score: {row['retrieval_score']:.4f}")
    print(f"Passage: {row['text'][:300]}")
    print("-" * 70)
```

```python
print("CONFIGURATION C — DENSE RETRIEVAL METRICS")
print("=" * 65)

print("\nOVERALL PERFORMANCE")

print(
    "Recall@5:",
    round(dense_recall_evaluation_df["recall_at_5"].mean(), 4)
)

print(
    "MRR@10:",
    round(dense_mrr_evaluation_df["reciprocal_rank_at_10"].mean(), 4)
)

print("\nCATEGORY-LEVEL PERFORMANCE")

recall_by_category = (
    dense_recall_evaluation_df
    .groupby("category")["recall_at_5"]
    .mean()
)

mrr_by_category = (
    dense_mrr_evaluation_df
    .groupby("category")["reciprocal_rank_at_10"]
    .mean()
)

category_metrics = pd.concat(
    [
        recall_by_category.rename("Recall@5"),
        mrr_by_category.rename("MRR@10")
    ],
    axis=1
)

display(category_metrics.round(4))
```

```python
def answer_with_dense(question, top_k=3):

    # Retrieve relevant research passages
    results = dense_search(question, top_k=top_k)

    sources = []
    evidence = []

    for i, (_, row) in enumerate(results.iterrows(), start=1):

        article_id = row["article_id"]
        title = row["title"]
        page = row["page_number"]
        passage = row["text"]

        citation = f"[{article_id}, p. {page}]"

        sources.append({
            "reference": citation,
            "article_id": article_id,
            "title": title,
            "page": page,
            "text": passage
        })

        evidence.append(
            f"Source {i}: {citation}\n"
            f"Article title: {title}\n"
            f"Evidence: {passage}"
        )

    # Abstain if no evidence is retrieved
    if not evidence:
        return {
            "answer": "The available research evidence is insufficient to answer this question.",
            "sources": sources
        }

    context = "\n\n".join(evidence)

    # Generate evidence-grounded answer with explicit citations
    response = llm.create_chat_completion(
        messages=[
            {
                "role": "system",
                "content": (
                    "You are an academic research assistant. "
                    "Answer using ONLY the supplied research passages. "
                    "Every factual claim must have an explicit citation "
                    "using the exact format [Article_ID, p. X]. "
                    "Use only the article IDs and page numbers provided. "
                    "Never invent statistics, percentages, patient counts, "
                    "clinical outcomes, or other information. "
                    "Do not confuse classification accuracy with patient "
                    "outcomes, treatment benefits, or satisfaction. "
                    "If the passages do not directly support an answer, "
                    "clearly state that the collection does not report "
                    "the requested information. "
                    "Do not create or guess citations."
                )
            },
            {
                "role": "user",
                "content": (
                    f"Research evidence:\n{context}\n\n"
                    f"Question: {question}\n\n"
                    "Provide a concise answer with supporting citations. "
                    "If evidence is insufficient, abstain."
                )
            }
        ],
        max_tokens=250,
        temperature=0.0
    )

    answer = response["choices"][0]["message"]["content"].strip()

    return {
        "answer": answer,
        "sources": sources
    }


print("Configuration C citation-improved function created successfully!")
```

```python
def dense_search(query, top_k=5):

    query_embedding = embedding_model_eval.encode(
        [query],
        convert_to_numpy=True,
        normalize_embeddings=True
    )

    scores = cosine_similarity(
        query_embedding,
        dense_document_embeddings
    )[0]

    top_indices = np.argsort(scores)[::-1][:top_k]

    # Use the full corpus, not the saved 400-row results.
    results = bm25_retrieval_df.iloc[top_indices].copy()

    results["retrieval_score"] = scores[top_indices]
    results["rank"] = range(1, len(results) + 1)

    return results


print("Dense search function corrected successfully!")
```

```python
def dense_search_citation(query, top_k=5):

    query_embedding = embedding_model_eval.encode(
        [query],
        convert_to_numpy=True,
        normalize_embeddings=True
    )

    scores = cosine_similarity(
        query_embedding,
        dense_citation_embeddings
    )[0]

    top_indices = np.argsort(scores)[::-1][:top_k]

    results = dense_citation_corpus.iloc[top_indices].copy()

    results["retrieval_score"] = scores[top_indices]
    results["rank"] = range(1, len(results) + 1)

    return results


print("Citation-improved dense search created successfully!")
```

```python
def answer_with_dense_citation(question, top_k=3):

    results = dense_search_citation(question, top_k=top_k)

    sources = []
    evidence = []

    for _, row in results.iterrows():

        article_id = row["article_id"]
        page = row["page_number"]

        reference = f"[{article_id}, p. {page}]"

        sources.append({
            "reference": reference,
            "article_id": article_id,
            "title": row["title"],
            "page": page,
            "text": row["text"]
        })

        evidence.append(
            f"{reference} {row['title']}: {row['text']}"
        )

    if not evidence:
        return {
            "answer": "Insufficient research evidence was retrieved.",
            "sources": sources
        }

    context = "\n\n".join(evidence)

    response = llm.create_chat_completion(
        messages=[
            {
                "role": "system",
                "content": (
                    "You are an academic research assistant. "
                    "Answer using only the supplied research passages. "
                    "Cite supporting claims using the exact format "
                    "[Article_ID, p. X]. "
                    "Never invent statistics, findings, or citations. "
                    "If the passages do not contain the requested information, "
                    "clearly state that the information is not reported. "
                    "Do not treat reference lists as research findings."
                )
            },
            {
                "role": "user",
                "content": (
                    f"Research evidence:\n{context}\n\n"
                    f"Question: {question}\n\n"
                    "Provide a concise, evidence-grounded answer."
                )
            }
        ],
        max_tokens=250,
        temperature=0.0
    )

    answer = response["choices"][0]["message"]["content"].strip()

    return {
        "answer": answer,
        "sources": sources
    }


print("Citation-improved answer function created successfully!")
```

```python
import re

def validate_percentage_claim(answer, sources):

    evidence_text = " ".join(
        source["text"] for source in sources
    )

    answer_percentages = set(
        re.findall(r"\b\d+(?:\.\d+)?\s*%", answer)
    )

    evidence_percentages = set(
        re.findall(r"\b\d+(?:\.\d+)?\s*%", evidence_text)
    )

    unsupported = answer_percentages - evidence_percentages

    if unsupported:
        return (
            "The retrieved research passages do not report "
            "a specific percentage of patients experiencing "
            "greater satisfaction with AI-based brain tumour diagnosis."
        )

    return answer


print("Percentage validation function created successfully!")
```

```python
def answer_with_dense_citation_final(question, top_k=3):

    results = dense_search_citation(question, top_k=top_k)

    sources = []
    evidence = []

    for _, row in results.iterrows():

        reference = f"[{row['article_id']}, p. {row['page_number']}]"

        sources.append({
            "reference": reference,
            "article_id": row["article_id"],
            "title": row["title"],
            "page": row["page_number"],
            "text": row["text"]
        })

        evidence.append(
            f"{reference}\n{row['text']}"
        )

    if not evidence:
        return {
            "answer": "Insufficient research evidence was retrieved.",
            "sources": sources
        }

    context = "\n\n".join(evidence)

    response = llm.create_chat_completion(
        messages=[
            {
                "role": "system",
                "content": (
                    "You are an academic research assistant. "
                    "Answer the question using only the supplied research passages. "
                    "Give a direct, concise answer in your own words. "
                    "Extract specific methods, architectures, findings, "
                    "or statistics only when explicitly supported. "
                    "Do not copy entire passages. "
                    "Do not invent information. "
                    "If the evidence is insufficient, say so clearly. "
                    "Return only the answer text, without citations."
                )
            },
            {
                "role": "user",
                "content": (
                    f"Research evidence:\n{context}\n\n"
                    f"Question: {question}\n\n"
                    "Answer in 2-4 sentences."
                )
            }
        ],
        temperature=0.0,
        max_tokens=200
    )

    answer = response["choices"][0]["message"]["content"].strip()

    answer = validate_percentage_claim(answer, sources)

    # Attach retrieved source references separately.
    citation_text = " ".join(
        source["reference"] for source in sources
    )

    return {
        "answer": answer,
        "citations": citation_text,
        "sources": sources
    }
```

```python
import numpy as np
from sklearn.metrics.pairwise import cosine_similarity

# BM25 retrieval
def hybrid_bm25_search(query, top_k=10):

    tokenized_query = query.lower().split()

    scores = hybrid_bm25_index.get_scores(tokenized_query)

    top_indices = np.argsort(scores)[::-1][:top_k]

    results = hybrid_corpus.iloc[top_indices].copy()

    results["bm25_score"] = scores[top_indices]
    results["bm25_rank"] = range(1, len(results) + 1)

    return results


# Dense retrieval
def hybrid_dense_search(query, top_k=10):

    query_embedding = embedding_model_eval.encode(
        [query],
        convert_to_numpy=True,
        normalize_embeddings=True
    )

    scores = cosine_similarity(
        query_embedding,
        dense_citation_embeddings
    )[0]

    top_indices = np.argsort(scores)[::-1][:top_k]

    results = hybrid_corpus.iloc[top_indices].copy()

    results["dense_score"] = scores[top_indices]
    results["dense_rank"] = range(1, len(results) + 1)

    return results


print("BM25 retrieval function created.")
print("Dense retrieval function created.")
```

```python
# Test both retrieval methods using Q001

test_question = evaluation_df.loc[
    evaluation_df["question_id"] == "Q001", "question"
].iloc[0]

bm25_test_results = hybrid_bm25_search(test_question, top_k=10)

dense_test_results = hybrid_dense_search(test_question, top_k=10)

print("QUESTION:")
print(test_question)

print("\nBM25 TOP 5 RESULTS:")
display(
    bm25_test_results[
        ["chunk_id", "article_id", "title", "bm25_score", "text"]
    ].head(5)
)

print("\nDENSE TOP 5 RESULTS:")
display(
    dense_test_results[
        ["chunk_id", "article_id", "title", "dense_score", "text"]
    ].head(5)
)
```

```python
# Reciprocal Rank Fusion

def reciprocal_rank_fusion(
    bm25_results,
    dense_results,
    top_k=10,
    k=60
):
    rank_scores = {}

    # Add BM25 ranking scores
    for _, row in bm25_results.iterrows():
        chunk_id = row["chunk_id"]
        rank_scores[chunk_id] = rank_scores.get(chunk_id, 0) + (
            1 / (k + row["bm25_rank"])
        )

    # Add dense ranking scores
    for _, row in dense_results.iterrows():
        chunk_id = row["chunk_id"]
        rank_scores[chunk_id] = rank_scores.get(chunk_id, 0) + (
            1 / (k + row["dense_rank"])
        )

    # Sort by combined RRF score
    ranked_ids = sorted(
        rank_scores,
        key=rank_scores.get,
        reverse=True
    )[:top_k]

    # Retrieve matching passages
    results = hybrid_corpus[
        hybrid_corpus["chunk_id"].isin(ranked_ids)
    ].copy()

    results["rrf_score"] = results["chunk_id"].map(rank_scores)

    results = results.sort_values(
        "rrf_score",
        ascending=False
    ).reset_index(drop=True)

    results["fusion_rank"] = range(1, len(results) + 1)

    return results
```

```python
hybrid_test_results = reciprocal_rank_fusion(
    bm25_test_results,
    dense_test_results,
    top_k=10
)

display(
    hybrid_test_results[
        ["chunk_id", "article_id", "title", "rrf_score", "fusion_rank"]
    ]
)
```

```python
import re
import json
import ast

citation_results = []

for _, row in configuration_b_results_400_df.iterrows():

    answer = str(row["generated_answer"])
    source_map = row["source_map"]

    if isinstance(source_map, str):
        try:
            source_map = json.loads(source_map)
        except:
            try:
                source_map = ast.literal_eval(source_map)
            except:
                source_map = {}

    if not isinstance(source_map, dict):
        source_map = {}

    cited_labels = re.findall(r"\[S\d+\]", answer)
    unique_labels = set(cited_labels)

    resolved_labels = {
        label for label in unique_labels
        if label[1:-1] in source_map
    }

    citation_results.append({
        "question_id": row["question_id"],
        "citation_count": len(unique_labels),
        "resolved_count": len(resolved_labels),
        "all_labels_resolve": (
            len(unique_labels) > 0
            and len(unique_labels) == len(resolved_labels)
        )
    })

citation_resolution_df = pd.DataFrame(citation_results)

print("Citation-label resolution results")
print("=" * 55)
print("Questions:", len(citation_resolution_df))
print("Questions with citations:",
      (citation_resolution_df["citation_count"] > 0).sum())
print("Questions with all labels resolved:",
      citation_resolution_df["all_labels_resolve"].sum())
print("Resolution percentage:",
      f"{citation_resolution_df['all_labels_resolve'].mean() * 100:.2f}%")
```

```python
def reciprocal_rank_fusion(bm25_results, dense_results, top_k=10, k=60):
    rank_scores = {}

    for _, row in bm25_results.iterrows():
        chunk_id = row["chunk_id"]
        rank_scores[chunk_id] = rank_scores.get(chunk_id, 0) + (
            1 / (k + row["bm25_rank"])
        )

    for _, row in dense_results.iterrows():
        chunk_id = row["chunk_id"]
        rank_scores[chunk_id] = rank_scores.get(chunk_id, 0) + (
            1 / (k + row["dense_rank"])
        )

    ranked_ids = sorted(
        rank_scores,
        key=rank_scores.get,
        reverse=True
    )[:top_k]

    results = hybrid_corpus[
        hybrid_corpus["chunk_id"].isin(ranked_ids)
    ].copy()

    results["rrf_score"] = results["chunk_id"].map(rank_scores)

    results = results.sort_values(
        "rrf_score",
        ascending=False
    ).reset_index(drop=True)

    results["fusion_rank"] = range(1, len(results) + 1)

    return results
```

```python
def retrieve_hybrid_evidence(query, top_k=5, retrieval_k=10):

    # Step 1: BM25 retrieval
    bm25_results = hybrid_bm25_search(
        query, top_k=retrieval_k
    )

    # Step 2: Dense retrieval
    dense_results = hybrid_dense_search(
        query, top_k=retrieval_k
    )

    # Step 3: Reciprocal Rank Fusion
    fused_results = reciprocal_rank_fusion(
        bm25_results,
        dense_results,
        top_k=retrieval_k
    )

    # Step 4: Cross-Encoder reranking
    pairs = [
        (query, passage)
        for passage in fused_results["text"].tolist()
    ]

    scores = cross_encoder.predict(pairs)

    fused_results = fused_results.copy()
    fused_results["rerank_score"] = scores

    reranked = fused_results.sort_values(
        "rerank_score",
        ascending=False
    ).reset_index(drop=True)

    reranked["rerank_rank"] = range(
        1, len(reranked) + 1
    )

    return reranked.head(top_k)
```

```python
def generate_hybrid_answer(question, selected_evidence, max_tokens=220):
    if selected_evidence is None or selected_evidence.empty:
        return {
            "answer": "Insufficient evidence in the retrieved passages.",
            "sources": {},
            "generation_time": 0,
            "finish_reason": "no_evidence",
            "prompt_tokens": 0
        }

    evidence_blocks = []
    source_mapping = {}

    for index, (_, row) in enumerate(selected_evidence.iterrows(), start=1):
        source_label = f"S{index}"

        source_mapping[source_label] = {
            "article_id": row["article_id"],
            "title": row["title"],
            "page_number": row["page_number"],
            "chunk_id": row["chunk_id"]
        }

        passage = " ".join(str(row["text"]).split()[:100])

        evidence_blocks.append(
            f"[{source_label}]\n"
            f"Article: {row['title']}\n"
            f"Page: {row['page_number']}\n"
            f"Passage: {passage}"
        )

    evidence_text = "\n\n".join(evidence_blocks)

    prompt = f"""
You are an academic research assistant.

Answer the question using only the supplied research evidence.

Rules:
1. Give a direct and relevant answer.
2. Include important information from the evidence.
3. Do not invent facts, numbers, or explanations.
4. Cite supporting passages using [S1], [S2], etc.
5. If the evidence does not answer the question,
   state: Insufficient evidence in the provided collection.
6. Do not guess when information is missing.

QUESTION: {question}

RESEARCH EVIDENCE: {evidence_text}

ANSWER:
"""

    prompt_tokens = len(llm.tokenize(prompt.encode("utf-8")))

    if prompt_tokens + max_tokens > llm.n_ctx():
        raise ValueError(
            f"Prompt requires {prompt_tokens} tokens; "
            f"only {llm.n_ctx() - max_tokens} are available."
        )

    start_time = time.time()

    response = llm(
        prompt,
        max_tokens=max_tokens,
        temperature=0.1,
        top_p=0.9,
        stop=["</s>"]
    )

    generation_time = time.time() - start_time

    return {
        "answer": response["choices"][0]["text"].strip(),
        "sources": source_mapping,
        "generation_time": generation_time,
        "finish_reason": response["choices"][0]["finish_reason"],
        "prompt_tokens": prompt_tokens
    }
```

```python

def generate_hybrid_answer(question, selected_evidence, max_tokens=220):

    if selected_evidence is None or selected_evidence.empty:
        return {
            "answer": "Insufficient evidence in the provided collection.",
            "sources": {},
            "generation_time": 0,
            "finish_reason": "no_evidence",
            "prompt_tokens": 0
        }

    evidence_blocks = []
    source_mapping = {}

    # Prepare the five evidence passages and citation labels
    for index, (_, row) in enumerate(
        selected_evidence.iterrows(), start=1
    ):
        source_label = f"S{index}"

        source_mapping[source_label] = {
            "article_id": row["article_id"],
            "title": row["title"],
            "page_number": row["page_number"],
            "chunk_id": row["chunk_id"]
        }

        passage = " ".join(str(row["text"]).split()[:100])

        evidence_blocks.append(
            f"[{source_label}]\n"
            f"Article: {row['title']}\n"
            f"Page: {row['page_number']}\n"
            f"Passage: {passage}"
        )

    evidence_text = "\n\n".join(evidence_blocks)

    # Construct the grounded answer prompt
    valid_labels = ", ".join(
        f"[{label}]" for label in source_mapping
    )

    prompt = f"""
You are an academic research assistant.

Answer the question using only the supplied research evidence.

Rules:
1. Give a direct answer in 2-4 sentences.
2. Use only information supported by the evidence.
3. Cite supporting evidence using only these labels: {valid_labels}.
4. Never invent or use citation labels that are not supplied.
5. Do not create a reference list.
6. If the evidence cannot answer the question, state:
   Insufficient evidence in the provided collection.
7. Do not guess or invent information.

QUESTION:
{question}

RESEARCH EVIDENCE:
{evidence_text}

ANSWER:
"""

    # Check the prompt against the model context window
    prompt_tokens = len(llm.tokenize(prompt.encode("utf-8")))

    if prompt_tokens + max_tokens > llm.n_ctx():
        raise ValueError(
            f"Prompt requires {prompt_tokens} tokens. "
            f"Available space for prompt: {llm.n_ctx() - max_tokens}."
        )

    # Generate answer
    start_time = time.time()

    response = llm(
        prompt,
        max_tokens=max_tokens,
        temperature=0.1,
        top_p=0.9,
        stop=["</s>"]
    )

    generation_time = time.time() - start_time

    answer = response["choices"][0]["text"].strip()
    finish_reason = response["choices"][0]["finish_reason"]

    return {
        "answer": answer,
        "sources": source_mapping,
        "generation_time": generation_time,
        "finish_reason": finish_reason,
        "prompt_tokens": prompt_tokens
    }
```

```python
q001_d_answer_v4 = generate_hybrid_answer(
    question=q001,
    selected_evidence=q001_d_evidence,
    max_tokens=220
)

print("Generated answer:")
print(q001_d_answer_v4["answer"])

print("\nPrompt tokens:", q001_d_answer_v4["prompt_tokens"])
print("Source count:", len(q001_d_answer_v4["sources"]))
print("Generation time:", q001_d_answer_v4["generation_time"])
print("Finish reason:", q001_d_answer_v4["finish_reason"])
```

```python
import pandas as pd
import json
from pathlib import Path

# Store Configuration D results separately
d_generation_records = []

# Output file
output_dir = Path("evaluation")
output_dir.mkdir(exist_ok=True)

output_path = output_dir / "configuration_D_llm_results.csv"

# Generate answers for all 80 questions
for index, row in df.iterrows():

    question_id = row["question_id"]
    question = row["question"]

    print(f"Processing {index + 1}/80: {question_id}")

    # Retrieve top 5 hybrid and reranked passages
    evidence = retrieve_hybrid_evidence(
        query=question,
        top_k=5,
        retrieval_k=10
    )

    try:
        # Generate a grounded answer
        result = generate_hybrid_answer(
            question=question,
            selected_evidence=evidence,
            max_tokens=220
        )

        record = {
            "question_id": question_id,
            "category": row["category"],
            "question": question,
            "generated_answer": result["answer"],
            "response_time": result["generation_time"],
            "finish_reason": result["finish_reason"],
            "prompt_tokens": result["prompt_tokens"],
            "retrieved_source_count": len(evidence),
            "source_map": json.dumps(
                result["sources"],
                ensure_ascii=False
            )
        }

    except Exception as e:
        print(f"Error processing {question_id}: {e}")

        record = {
            "question_id": question_id,
            "category": row["category"],
            "question": question,
            "generated_answer": None,
            "response_time": None,
            "finish_reason": "error",
            "prompt_tokens": None,
            "retrieved_source_count": len(evidence),
            "source_map": None
        }

    d_generation_records.append(record)

    # Save a checkpoint every 10 questions
    if (index + 1) % 10 == 0:
        checkpoint_df = pd.DataFrame(d_generation_records)
        checkpoint_df.to_csv(output_path, index=False)
        print(f"Checkpoint saved: {index + 1}/80")

# Final save
configuration_d_llm_results = pd.DataFrame(d_generation_records)

configuration_d_llm_results.to_csv(
    output_path,
    index=False
)

print("\nConfiguration D answer generation completed.")
print("Dataset shape:", configuration_d_llm_results.shape)
print("Saved to:", output_path)
print("File exists:", output_path.exists())
```

# Create an improved generation function

```python
def generate_hybrid_answer_v2(question, selected_evidence, max_tokens=160):

    if selected_evidence is None or selected_evidence.empty:
        return {
            "answer": "Insufficient evidence in the provided collection.",
            "sources": {},
            "generation_time": 0,
            "finish_reason": "no_evidence",
            "prompt_tokens": 0
        }

    evidence_blocks = []
    source_mapping = {}

    for index, (_, row) in enumerate(
        selected_evidence.iterrows(), start=1
    ):
        source_label = f"S{index}"

        source_mapping[source_label] = {
            "article_id": row["article_id"],
            "title": row["title"],
            "page_number": row["page_number"],
            "chunk_id": row["chunk_id"]
        }

        passage = " ".join(str(row["text"]).split()[:100])

        evidence_blocks.append(
            f"[{source_label}] {passage}"
        )

    evidence_text = "\n\n".join(evidence_blocks)

    valid_labels = ", ".join(
        f"[{label}]" for label in source_mapping
    )

    prompt = f"""
You are a precise academic research assistant.

Answer the question using only the research evidence below.

STRICT RULES:
1. Give one concise answer in 2-3 sentences.
2. Answer only what the question asks.
3. Do not add general background or unsupported explanations.
4. Do not repeat information.
5. Cite only passages that directly support your statements.
6. Use citation labels exactly as provided: {valid_labels}.
7. Make one final decision:
   - If the evidence answers the question, provide the answer with citations.
   - If the evidence does not answer the question, write exactly:
     Insufficient evidence in the provided collection.
8. Never combine an answer with an insufficient-evidence statement.
9. Do not generate a reference list.
10. Stop immediately after completing the answer.

QUESTION:
{question}

EVIDENCE:
{evidence_text}

FINAL ANSWER:
"""

    prompt_tokens = len(llm.tokenize(prompt.encode("utf-8")))

    if prompt_tokens + max_tokens > llm.n_ctx():
        raise ValueError(
            f"Prompt requires {prompt_tokens} tokens. "
            f"Available space: {llm.n_ctx() - max_tokens}."
        )

    start_time = time.time()

    response = llm(
        prompt,
        max_tokens=max_tokens,
        temperature=0.0,
        top_p=0.9,
        stop=["</s>", "\n\n"]
    )

    generation_time = time.time() - start_time

    answer = response["choices"][0]["text"].strip()

    return {
        "answer": answer,
        "sources": source_mapping,
        "generation_time": generation_time,
        "finish_reason": response["choices"][0]["finish_reason"],
        "prompt_tokens": prompt_tokens
    }
```

# Test a less restrictive prompt

```python
def generate_hybrid_answer_v3(question, selected_evidence, max_tokens=160):

    if selected_evidence is None or selected_evidence.empty:
        return {
            "answer": "Insufficient evidence in the provided collection.",
            "sources": {},
            "generation_time": 0,
            "finish_reason": "no_evidence",
            "prompt_tokens": 0
        }

    evidence_blocks = []
    source_mapping = {}

    for index, (_, row) in enumerate(
        selected_evidence.iterrows(), start=1
    ):
        source_label = f"S{index}"

        source_mapping[source_label] = {
            "article_id": row["article_id"],
            "title": row["title"],
            "page_number": row["page_number"],
            "chunk_id": row["chunk_id"]
        }

        passage = " ".join(str(row["text"]).split()[:100])

        evidence_blocks.append(
            f"[{source_label}] {passage}"
        )

    evidence_text = "\n\n".join(evidence_blocks)

    valid_labels = ", ".join(
        f"[{label}]" for label in source_mapping
    )

    prompt = f"""
You are an academic research assistant.

Answer the question using the research passages provided.

Instructions:
1. Identify the information that directly answers the question.
2. Give a clear answer in 2-3 sentences.
3. Include specific methods, names or figures when available.
4. Combine relevant information from different passages when useful.
5. Do not add information that is not supported by the passages.
6. Add citations immediately after the supported statements.
7. Use only these citation labels: {valid_labels}.
8. Do not repeat statements or citations.
9. Only state "Insufficient evidence in the provided collection"
   if none of the passages contains information that answers the question.
10. Stop after giving the answer.

QUESTION:
{question}

RESEARCH PASSAGES:
{evidence_text}

ANSWER:
"""

    prompt_tokens = len(llm.tokenize(prompt.encode("utf-8")))

    if prompt_tokens + max_tokens > llm.n_ctx():
        raise ValueError(
            f"Prompt requires {prompt_tokens} tokens. "
            f"Available space: {llm.n_ctx() - max_tokens}."
        )

    start_time = time.time()

    response = llm(
        prompt,
        max_tokens=max_tokens,
        temperature=0.1,
        top_p=0.9,
        stop=["</s>", "\n\n"]
    )

    generation_time = time.time() - start_time

    return {
        "answer": response["choices"][0]["text"].strip(),
        "sources": source_mapping,
        "generation_time": generation_time,
        "finish_reason": response["choices"][0]["finish_reason"],
        "prompt_tokens": prompt_tokens
    }
```

```python

# Test a more focused prompt using the same three complete passages.

def generate_hybrid_answer_focused(question, selected_evidence, max_tokens=220):

    evidence_blocks = []
    source_mapping = {}

    for index, (_, row) in enumerate(selected_evidence.iterrows(), start=1):
        label = f"S{index}"

        source_mapping[label] = {
            "article_id": row["article_id"],
            "title": row["title"],
            "page_number": row["page_number"],
            "chunk_id": row["chunk_id"]
        }

        evidence_blocks.append(
            f"[{label}] {row['text']}"
        )

    evidence_text = "\n\n".join(evidence_blocks)

    prompt = f"""You are an academic research assistant.

Answer the question using only the evidence.

Give a concise answer in 2-3 sentences.
Mention the specific methods that answer the question.
Put the relevant source citation immediately after each claim.
Use only citations [S1], [S2], or [S3].
Do not repeat the question or add unrelated background.
Do not invent information.

Question: {question}

Evidence:
{evidence_text}

Answer:
"""

    prompt_tokens = len(llm.tokenize(prompt.encode("utf-8")))

    if prompt_tokens + max_tokens > llm.n_ctx():
        raise ValueError(
            f"Prompt requires {prompt_tokens} tokens; "
            f"only {llm.n_ctx() - max_tokens} are available."
        )

    start_time = time.time()

    response = llm(
        prompt,
        max_tokens=max_tokens,
        temperature=0.1,
        top_p=0.9,
        stop=["</s>"]
    )

    return {
        "answer": response["choices"][0]["text"].strip(),
        "sources": source_mapping,
        "generation_time": time.time() - start_time,
        "finish_reason": response["choices"][0]["finish_reason"],
        "prompt_tokens": prompt_tokens
    }


q001_focused_test = generate_hybrid_answer_focused(
    q001_question,
    q001_d_evidence.head(3)
)

print("GENERATED ANSWER:")
print(q001_focused_test["answer"])
print("\nPROMPT TOKENS:", q001_focused_test["prompt_tokens"])
print("GENERATION TIME:", round(q001_focused_test["generation_time"], 3))
print("FINISH REASON:", q001_focused_test["finish_reason"])
```

```python
import json
import re
import pandas as pd
from pathlib import Path

# Create blank manual evaluation columns
for col in [
    "correctness_score",
    "relevance_score",
    "faithfulness_score",
    "citation_validity",
    "abstention_correct",
    "review_notes"
]:
    d_eval_df[col] = pd.NA

# Extract citation labels and check whether they resolve
d_eval_df["citation_labels"] = d_eval_df["generated_answer"].apply(
    lambda answer: sorted(set(re.findall(r"\[S\d+\]", str(answer))))
)

def check_citations(row):
    try:
        source_map = (
            json.loads(row["source_map"])
            if isinstance(row["source_map"], str)
            else row["source_map"]
        )
        source_map = source_map or {}
    except (json.JSONDecodeError, TypeError):
        source_map = {}

    labels = row["citation_labels"]
    resolved = [label for label in labels if label.strip("[]") in source_map]

    return pd.Series({
        "citation_count": len(labels),
        "resolved_citation_count": len(resolved),
        "citation_check": (
            "All labels resolve" if len(labels) == len(resolved)
            else "Some labels unresolved"
        )
    })

d_eval_df[
    ["citation_count", "resolved_citation_count", "citation_check"]
] = d_eval_df.apply(check_citations, axis=1)

# Save separately; original D results remain unchanged
d_eval_path = Path("evaluation/configuration_D_evaluation_working.csv")
d_eval_df.to_csv(d_eval_path, index=False)

print("Saved:", d_eval_path)
print("File exists:", d_eval_path.exists())
print("Shape:", d_eval_df.shape)

print("\nCitation-label check:")
print(d_eval_df["citation_check"].value_counts(dropna=False))

print("\nEvaluation columns:")
print(d_eval_df.columns.tolist())
```

```python
import pandas as pd
from pathlib import Path

# 1. Check that all 80 questions have been scored
score_columns = [
    "correctness_score",
    "relevance_score",
    "faithfulness_score",
    "citation_validity"
]

print("EVALUATION COMPLETENESS")
print("=" * 55)
print("Total questions:", len(d_eval_df))

for col in score_columns:
    print(f"{col}: {d_eval_df[col].notna().sum()}/80 scored")

print("\nUnanswered evaluation fields:")
print(d_eval_df[score_columns].isna().sum())

# 2. Calculate answer-quality metrics
print("\nCONFIGURATION D ANSWER-QUALITY METRICS")
print("=" * 55)

for col in score_columns[:3]:
    percentage = d_eval_df[col].mean() / 2 * 100
    print(f"{col}: {percentage:.2f}%")

citation_percentage = (
    d_eval_df["citation_validity"].mean() * 100
)
print(f"citation_validity: {citation_percentage:.2f}%")

# 3. Fully correct answers
fully_correct = (
    (d_eval_df["correctness_score"] == 2) &
    (d_eval_df["relevance_score"] == 2) &
    (d_eval_df["faithfulness_score"] == 2) &
    (d_eval_df["citation_validity"] == 1)
)

print(
    f"\nFully correct answers: {fully_correct.sum()}/80 "
    f"({fully_correct.mean() * 100:.2f}%)"
)

# 4. Abstention accuracy
unanswerable_df = d_eval_df[
    d_eval_df["category"] == "Unanswerable"
]

print("\nABSTENTION")
print("=" * 55)
print("Unanswerable questions:", len(unanswerable_df))
print(
    "Correct abstentions:",
    int(unanswerable_df["abstention_correct"].sum()),
    "/",
    len(unanswerable_df)
)
print(
    "Abstention accuracy:",
    f"{unanswerable_df['abstention_correct'].mean() * 100:.2f}%"
)

# 5. Response time
print("\nRESPONSE TIME")
print("=" * 55)
print("Mean:", f"{d_eval_df['response_time'].mean():.3f} seconds")
print("Median:", f"{d_eval_df['response_time'].median():.3f} seconds")

# 6. Inspect saved retrieval evaluation columns
retrieval_path = Path(
    "evaluation/configuration_D_retrieval_evaluation.csv"
)
d_retrieval_check = pd.read_csv(retrieval_path)

print("\nSAVED RETRIEVAL EVALUATION")
print("=" * 55)
print("File exists:", retrieval_path.exists())
print("Shape:", d_retrieval_check.shape)
print("Columns:", d_retrieval_check.columns.tolist())
```

```python
import json
import numpy as np
import pandas as pd

# 1. Verify retrieval metrics
print("RETRIEVAL METRICS")
print("=" * 55)

print(
    "Recall@5:",
    f"{d_retrieval_check['recall_at_5'].mean():.4f}"
)

print(
    "MRR@10:",
    f"{d_retrieval_check['reciprocal_rank_at_10'].mean():.4f}"
)

# 2. Verify citation-label resolution
def verify_resolution(row):
    labels = row["citation_labels"]

    source_map = (
        json.loads(row["source_map"])
        if isinstance(row["source_map"], str)
        else row["source_map"]
    )

    source_map = source_map or {}

    resolved = [
        label for label in labels
        if label.strip("[]") in source_map
    ]

    return len(labels), len(resolved)

resolution_results = d_eval_df.apply(
    verify_resolution, axis=1
)

total_labels = sum(x[0] for x in resolution_results)
resolved_labels = sum(x[1] for x in resolution_results)

print("\nCITATION-LABEL RESOLUTION")
print("=" * 55)
print("Total citation labels:", total_labels)
print("Resolved citation labels:", resolved_labels)

if total_labels > 0:
    print(
        "Label resolution rate:",
        f"{resolved_labels / total_labels * 100:.2f}%"
    )
else:
    print("No citation labels were generated.")

# 3. Check final evaluation file
final_path = Path(
    "evaluation/configuration_D_evaluation_working.csv"
)

print("\nFINAL EVALUATION FILE")
print("=" * 55)
print("File exists:", final_path.exists())
print("Rows:", len(pd.read_csv(final_path)))
```

```python

import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# ==========================================
# 1. ENTER THE COMPLETED EVALUATION RESULTS
# ==========================================

results = {
    "Configuration": ["A", "B", "C", "D"],

    "Correctness (%)": [
        42.50, 26.25, 35.00, 54.37
    ],

    "Relevance (%)": [
        98.75, 65.62, 93.75, 97.50
    ],

    "Faithfulness (%)": [
        40.00, 31.87, 39.38, 50.62
    ],

    "Citation Validity (%)": [
        np.nan, 36.25, 1.25, 11.25
    ],

    "Fully Correct (%)": [
        27.50, 15.00, 16.00, 6.25
    ],

    "Abstention Accuracy (%)": [
        55.00, 75.00, 10.00, 65.00
    ],

    "Mean Response Time (seconds)": [
        1.493, 8.150, 2.971, 4.870
    ],

    "Recall@5 (%)": [
        np.nan, 45.33, 33.75, 43.22
    ],

    "MRR@10 (%)": [
        np.nan, 42.03, 29.84, 45.53
    ]
}

df_comparison = pd.DataFrame(results)

# ==========================================
# 2. DISPLAY COMPARISON TABLE
# ==========================================

print("========== CONFIGURATION COMPARISON ==========")

display(
    df_comparison.set_index("Configuration").round(2)
)

# ==========================================
# 3. ANSWER QUALITY COMPARISON
# ==========================================

quality_metrics = [
    "Correctness (%)",
    "Relevance (%)",
    "Faithfulness (%)",
    "Citation Validity (%)"
]

ax = (
    df_comparison
    .set_index("Configuration")[quality_metrics]
    .plot(kind="bar", figsize=(12, 6), width=0.8)
)

ax.set_title("Answer Quality Comparison: A, B, C and D")
ax.set_xlabel("Configuration")
ax.set_ylabel("Score (%)")
ax.set_ylim(0, 110)
ax.legend(title="Evaluation Metrics", bbox_to_anchor=(1.02, 1))
ax.grid(axis="y", linestyle="--", alpha=0.4)

plt.xticks(rotation=0)
plt.tight_layout()
plt.show()

# ==========================================
# 4. FULLY CORRECT AND ABSTENTION COMPARISON
# ==========================================

plt.figure(figsize=(10, 5))

x = np.arange(len(df_comparison))
width = 0.35

plt.bar(
    x - width/2,
    df_comparison["Fully Correct (%)"],
    width,
    label="Fully Correct"
)

plt.bar(
    x + width/2,
    df_comparison["Abstention Accuracy (%)"],
    width,
    label="Abstention Accuracy"
)

plt.title("Fully Correct Answers vs Abstention Accuracy")
plt.xlabel("Configuration")
plt.ylabel("Score (%)")
plt.xticks(x, df_comparison["Configuration"])
plt.ylim(0, 100)
plt.legend()
plt.grid(axis="y", linestyle="--", alpha=0.4)

plt.tight_layout()
plt.show()

# ==========================================
# 5. RESPONSE TIME COMPARISON
# ==========================================

plt.figure(figsize=(9, 5))

plt.bar(
    df_comparison["Configuration"],
    df_comparison["Mean Response Time (seconds)"]
)

plt.title("Mean Response Time Comparison")
plt.xlabel("Configuration")
plt.ylabel("Response Time (seconds)")
plt.grid(axis="y", linestyle="--", alpha=0.4)

plt.tight_layout()
plt.show()

# ==========================================
# 6. RETRIEVAL PERFORMANCE COMPARISON
# ==========================================

retrieval_metrics = [
    "Recall@5 (%)",
    "MRR@10 (%)"
]

ax = (
    df_comparison
    .set_index("Configuration")[retrieval_metrics]
    .plot(kind="bar", figsize=(9, 5), width=0.7)
)

ax.set_title("Retrieval Performance: B, C and D")
ax.set_xlabel("Configuration")
ax.set_ylabel("Score (%)")
ax.set_ylim(0, 60)
ax.legend(title="Retrieval Metrics")
ax.grid(axis="y", linestyle="--", alpha=0.4)

plt.xticks(rotation=0)
plt.tight_layout()
plt.show()

# ==========================================
# 7. PRINT BEST CONFIGURATION PER METRIC
# ==========================================

print("\n========== BEST CONFIGURATION PER METRIC ==========")

for metric in [
    "Correctness (%)",
    "Relevance (%)",
    "Faithfulness (%)",
    "Citation Validity (%)",
    "Fully Correct (%)",
    "Abstention Accuracy (%)",
    "Recall@5 (%)",
    "MRR@10 (%)"
]:
    valid = df_comparison.dropna(subset=[metric])
    best = valid.loc[valid[metric].idxmax()]

    print(
        f"{metric}: Configuration "
        f"{best['Configuration']} ({best[metric]:.2f}%)"
    )

fastest = df_comparison.loc[
    df_comparison["Mean Response Time (seconds)"].idxmin()
]

print(
    "\nFastest configuration:",
    fastest["Configuration"],
    f"({fastest['Mean Response Time (seconds)']:.3f} seconds)"
)
```

```python

import matplotlib.pyplot as plt
import numpy as np

# Answer quality: all four configurations
metrics = [
    "Correctness (%)",
    "Relevance (%)",
    "Faithfulness (%)",
    "Citation Validity (%)"
]

for metric in metrics:
    plt.figure(figsize=(8, 5))

    values = df_comparison[metric]

    plt.bar(
        df_comparison["Configuration"],
        values
    )

    plt.title(f"{metric} - Configuration Comparison")
    plt.xlabel("Configuration")
    plt.ylabel("Score (%)")
    plt.ylim(0, 110)
    plt.grid(axis="y", alpha=0.3)

    for i, value in enumerate(values):
        label = "N/A" if pd.isna(value) else f"{value:.2f}%"
        if not pd.isna(value):
            plt.text(i, value + 2, label, ha="center")
        else:
            plt.text(i, 3, label, ha="center")

    plt.tight_layout()
    plt.show()


# Retrieval metrics: only B, C and D
retrieval_df = df_comparison.dropna(
    subset=["Recall@5 (%)", "MRR@10 (%)"]
)

retrieval_df.set_index("Configuration")[
    ["Recall@5 (%)", "MRR@10 (%)"]
].plot(
    kind="bar",
    figsize=(9, 5)
)

plt.title("Retrieval Performance (A does not use retrieval)")
plt.ylabel("Score (%)")
plt.xlabel("Configuration")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python

import pandas as pd
import numpy as np

# ==========================================
# 1. NUMERICAL COMPARISON
# ==========================================

comparison = df_comparison.copy()

# Set configuration as index
comparison = comparison.set_index("Configuration")

print("=" * 75)
print("NUMERICAL PERFORMANCE COMPARISON: A, B, C AND D")
print("=" * 75)

display(comparison.round(3))


# ==========================================
# 2. RANK EACH CONFIGURATION PER METRIC
# ==========================================

higher_is_better = [
    "Correctness (%)",
    "Relevance (%)",
    "Faithfulness (%)",
    "Citation Validity (%)",
    "Fully Correct (%)",
    "Abstention Accuracy (%)",
    "Recall@5 (%)",
    "MRR@10 (%)"
]

rankings = comparison[higher_is_better].rank(
    ascending=False,
    method="min"
)

print("\n========== PERFORMANCE RANKINGS ==========")
display(rankings)


# ==========================================
# 3. RESPONSE TIME RANKING
# ==========================================

response_ranking = comparison[
    ["Mean Response Time (seconds)"]
].rank(ascending=True, method="min")

print("\n========== RESPONSE TIME RANKING ==========")
display(response_ranking)


# ==========================================
# 4. OVERALL RANKING
# ==========================================

# Average rank across applicable metrics.
# Missing metrics (N/A) are excluded.
overall_ranking = rankings.mean(axis=1).sort_values()

print("\n========== OVERALL AVERAGE RANK ==========")
display(
    overall_ranking.to_frame(
        name="Average Rank (lower is better)"
    ).round(2)
)


# ==========================================
# 5. BEST CONFIGURATION PER METRIC
# ==========================================

print("\n========== BEST CONFIGURATION PER METRIC ==========")

for metric in higher_is_better:
    valid_scores = comparison[metric].dropna()

    best_config = valid_scores.idxmax()
    best_score = valid_scores.max()

    print(
        f"{metric:<30} "
        f"Configuration {best_config}: {best_score:.2f}%"
    )

fastest = comparison[
    "Mean Response Time (seconds)"
].idxmin()

print(
    f"\nFastest configuration: {fastest} "
    f"({comparison.loc[fastest, 'Mean Response Time (seconds)']:.3f} seconds)"
)


# ==========================================
# 6. EXPORT NUMERICAL COMPARISON
# ==========================================

output_path = PROJECT_DIR / "configuration_comparison.csv"

comparison.round(3).to_csv(output_path)

print("\nComparison saved to:", output_path)
```

```python
import joblib
import numpy as np
import pandas as pd
from pathlib import Path

print("========== CONFIGURATION D ARTIFACT INSPECTION ==========")

artifact_files = {
    "Retrieval chunks": RETRIEVAL_DIR / "retrieval_chunks.pkl",
    "BM25 index": RETRIEVAL_DIR / "bm25_index.joblib",
    "Embeddings": RETRIEVAL_DIR / "document_embeddings.npy",
    "Retrieval config": RETRIEVAL_DIR / "retrieval_config.joblib"
}

for name, path in artifact_files.items():
    print(f"\n{name}")
    print("Exists:", path.exists())

    if not path.exists():
        continue

    if path.suffix == ".pkl":
        obj = pd.read_pickle(path)
        print("Type:", type(obj))
        print("Shape:", getattr(obj, "shape", "N/A"))
        print("Columns:", getattr(obj, "columns", "N/A"))

    elif path.suffix == ".npy":
        obj = np.load(path, mmap_mode="r")
        print("Shape:", obj.shape)
        print("Data type:", obj.dtype)

    else:
        obj = joblib.load(path)
        print("Type:", type(obj))

        if isinstance(obj, dict):
            print("Keys:", list(obj.keys()))
        elif hasattr(obj, "__dict__"):
            print("Attributes:", list(obj.__dict__.keys()))

print("\n========== END ==========")
```

```python
import joblib
import numpy as np
import json

from pathlib import Path

# Separate folder for the application
DEPLOY_DIR = PROJECT_DIR / "artifacts" / "deployment_D"
DEPLOY_DIR.mkdir(parents=True, exist_ok=True)

# Save the exact in-memory Configuration D objects
joblib.dump(
    hybrid_corpus,
    DEPLOY_DIR / "hybrid_corpus.joblib"
)

joblib.dump(
    hybrid_bm25_index,
    DEPLOY_DIR / "hybrid_bm25_index.joblib"
)

np.save(
    DEPLOY_DIR / "dense_embeddings.npy",
    dense_citation_embeddings
)

# Save model and retrieval configuration
deployment_config = {
    "embedding_model": "sentence-transformers/all-MiniLM-L6-v2",
    "cross_encoder_model": "cross-encoder/ms-marco-MiniLM-L-6-v2",
    "llm_model": "qwen2.5-0.5b-instruct-q4_k_m.gguf",
    "context_size": 2048,
    "total_chunks": len(hybrid_corpus),
    "unique_articles": int(hybrid_corpus["article_id"].nunique()),
    "retrieval_k": 10,
    "top_k": 5,
    "rrf_k": 60
}

with open(
    DEPLOY_DIR / "deployment_config.json",
    "w",
    encoding="utf-8"
) as file:
    json.dump(deployment_config, file, indent=4)

print("========== DEPLOYMENT ARTIFACTS SAVED ==========")

for path in sorted(DEPLOY_DIR.iterdir()):
    if path.is_file():
        print(f"{path.name}: {path.stat().st_size / (1024**2):.2f} MB")

print("\nCorpus rows:", len(hybrid_corpus))
print("BM25 corpus size:", hybrid_bm25_index.corpus_size)
print("Embedding shape:", dense_citation_embeddings.shape)

print("\nAll three components aligned:",
      len(hybrid_corpus) ==
      hybrid_bm25_index.corpus_size ==
      dense_citation_embeddings.shape[0])
```

```python
from pathlib import Path

PROJECT_DIR = Path("/Users/macbookair/DAT401_Research_Assistant")
app_path = PROJECT_DIR / "app.py"

# Read the incorrectly saved file
raw_code = app_path.read_text(encoding="utf-8")

# Find the beginning of the application code
start_marker = "app_code = r'''"
start = raw_code.find(start_marker)

if start == -1:
    raise ValueError("Could not find the application code.")

content_start = start + len(start_marker)

# Find the closing triple quotes
end = raw_code.rfind("'''")

if end <= content_start:
    raise ValueError("Could not find the closing triple quotes.")

# Extract only the Streamlit application
streamlit_code = raw_code[content_start:end].lstrip("\n")

# Save the corrected application
app_path.write_text(streamlit_code, encoding="utf-8")

print("Application corrected and saved!")
print("File size:", app_path.stat().st_size, "bytes")
print("\nFirst 5 lines:")

print("\n".join(streamlit_code.splitlines()[:5]))
```
